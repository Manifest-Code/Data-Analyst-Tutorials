## Q1 What is the difference between a function and a method in Python?  

- Function:    
  - A function is defined outside of any class.     
  - It is a reusable block of code that performs a specific task.     
  - You call it directly by its name and pass any necessary data as arguments.    
  - It is defined using the def keyword.  
  - It exists independently (Global scope).    
  - It is called by name: function_name(arguments)    
  example:  
   ```
     def add_numbers(a, b):  
        return a + b    
     print(add_numbers(5, 10)) # Calling the function  
                                                            
- Methods:  
  - A method is a function that is defined inside a class.  
  - It is used to define the behaviors of an object.  
  - The key characteristic of a method is that it implicitly acts on the data (attributes) contained within the object that calls it.  
  - It is defined inside a class using def    
  - It is associated with object or class    
  - Called on an object using dot notation: object.method_name().  
  - Methods almost always take self as the first argument, which represents the specific instance of the object.  
  Example:  
  ```
     class Calculator:  
         def __init__(self, value):  
              self.value = value  

         def add_to_value(self, amount):  # This is a method  
           self.value += amount  
           return self.value  

         my_calc = Calculator(10)  
         print(my_calc.add_to_value(5)) # Calling the method on the object  
                                                                      

## Q2 Explain the concept of function arguments and parameters in Python.  

- You can think as a parameter is a "placeholder" or a variable name, and an argument as the "actual value" you fill that placeholder with.    
- Parameters:  
   - Parameters are the variables listed inside the parentheses in the function definition.  
   - They act as containers for the data the function expects to receive.    
- Arguments :  
   - Arguments are the actual values sent to the function when it is called.  
example:  
     ```
       # 'name' is the PARAMETER (the placeholder) 
       def greet(name):
         print(f"Hello, {name}!")

       # "Alice" is the ARGUMENT (the actual data)
       greet("Alice")
                                                    

- Types of Arguments in Python:  
   1. Positional Arguments:  
      - These are the most common.  
      - The values are assigned to parameters based on their order.  
      Example:  
        ```  
          def addition(a, b):  
             return a + b  
          addition("Data","Analyst")
                                     

   2. Keyword Arguments:  
      -  You can explicitly state which parameter you are providing a value for.  
      -  This means the order doesn't matter.  
      Example:  
         ```
            def greet(name, message):  
              print(f"{message}, {name}")  
	         greet(message="Hello", name = "Alice")  

    3. Default Parameters:  
       - You can provide a "fallback" value in case an argument isn't provided during the function call.  
        Example:  
          ``` 
            def sum1(b, c, a=0):    
              return a + b     
	         sum1(1,2)  
                      

    4. Arbitrary Arguments (*args and **kwargs):    
       - Sometimes you don't know how many arguments will be passed.    
       - *args: Receives a tuple of extra positional arguments.  
       - **kwargs: Receives a dictionary of extra keyword arguments.  
         Example:  
           ```  
            def test1(*arg, **arg1):    
              return arg, arg1  
	         test1(1, 2, 4, a=100, b=50)   
                                     

## Q3 What are the different ways to define and call a function in Python?  

1. Standard Function (def):  
- This is the most common way. It uses the def keyword, a unique name, and a block of indented code.  
  Example:  
  ```  
     def square(number):  
       return number * number 
     result = square(4)  # Returns 16 
                             
2. Lambda Functions (Anonymous function):  
- Lambda functions are small, anonymous functions defined using the lambda keyword.  
- They are used for creating small, throwaway functions without the need to formally define a function using def  
  Example:  
  ```  
    sq_lambda = lambda x: x**2  
    sq_lambda(2)   # returns 4  
                               
3. Functions with Default Parameters:  
-  You can define a function where some arguments are optional.  
-  If the user doesn't provide a value, Python uses the default value you specified.
   Example:  
   ```  
    def sum1(b, c, a=0):  
      return a + b + c
	  sum1(1,2)    # output 3  
                       
4. Map:     
- The map function applies a given function to all items in an input iterable (like a list) and returns an iterator with the results.  
- It is a function which executes a specified function for each of iterm of an iterable.  
  Example:  
  ```  
     l = [1, 2, 3, 4, 5]  
     def add(x):
       return x+10
     list(map(add, l))  
     #output - [1, 4, 9, 16, 25]    

     # using Lambda with Map
     list(map(lambda x: x**2, l))  
     #output - [1, 4, 9, 16, 25]    

5. Reduce Function:    
- It is also consider as folding / reduction of a function.
  - Reduce always take 2 argument function.  
  - It uses the result as the first argument for the next step until only one value remains.  
  - Example:  
     ```
      from functools import reduce
      l = [2, 1, 3, 4, 4, 5, 6]
      reduce(lambda x, y: x+y, l)
      #Output - 25     
                               
6. Filter Function:  
- It is used to filter elements from an iterable based on some condition.  
  Example:  
  ```  
   l = [2, 1, 3, 4, 4, 5, 6]  
   list(filter(lambda x: x%2 ==0, l))
   #output - [2, 4, 4, 6]

## Q5 What are iterators in Python and how do they differ from iterables?

       ```
         iter_list = iter(["Apple", "Mango", "Strawberry"])

## Q6  Explain the concept of generators in Python and how they are defined.    

- Definition: Generators are a simple way to create iterators using functions and the yield keyword instead of returning values.    

- Key Points:    
  - yield Keyword:    
    - Yield vs. Return: Unlike the return statement, which exits the function, yield returns a value and pauses the function’s state, allowing it    
      to resume from where it left off.    
  - Multiple Yields: Functions can yield multiple values over time, producing a sequence of results.    
  - On-Demand Data: Generators produce values one at a time and only when required, avoiding the need to generate or store all values upfront.  
  - Memory Efficiency: Since values are generated as needed and not all at once, generators use minimal memory.    
  - Generators return data in an iterable format, allowing the sequence to be iterated over, much like a list or tuple.    
  - Simpler Syntax: Using yield simplifies the creation of iterators, making the code more concise and readable.  
    Example:  
      ```
         def square_number_generators(n):
             for i in range(n):
             yield i ** 2
        
         gen = square_number_generators(10)
         gen

         next(gen)
         #output 0  

         next(gen)
         #output 1  ....
                   

## Q7 What are the advantages of using generators over regular functions?    

- Advantages of using generators over regular functions: 
  - Generator uses "yield" keyword which returns a value and pauses the function’s state, allowing it to resume from where it left off where as  
    Regular Function uses 'return' keyword which exits the function.   
  - Regular Function stores entire result in access memory usage but, Generators produce values one at a time and only when required, avoiding  
    the need to generate or store all values upfront. Since values are generated as needed and not all at once, generators use minimal memory.  
  - Regular function runs to completion, then returns. where as Generators pauses and resumes state.  
  - In a regular function, you have to wait for the entire loop to finish before you can process the first item. In a generator, you can start  
    processing the first item as soon as it is yielded.  
  - Which is why Regular functions are limited by available RAM and Generators are virtually unlimited (Lazy loading).  
  - Regular Function is used when if you need to access the data multiple times, sort it, or access items out of order (by index).
  - But if you are dealing with large datasets (files, database rows, network streams) or if you only need to iterate through the data  
    once then Generators are used.

## Q9 Explain the purpose and usage of the `map()` function in Python.    

- The map() function is a built-in Python tool that allows you to apply a specific function to every item in an iterable (like a list or tuple)  
  without using an explicit for loop.
- It is a pillar of "Functional Programming" in Python, designed to make code more concise and readable when performing bulk transformations.  
1. The Syntax
   - The map() function takes two primary arguments:  
     - Function: The logic you want to apply.
     - Iterable: The data you want to transform.  
     Example:  
     ```
        l = [1, 2, 3, 4, 5]
        list(map(lambda x: x**2, l)) 
2. map() does not return a list; it returns an "iterator" (a map object). This is memory efficient because it doesn't calculate the values until    
   you actually try to look at them.   
3. Lazy Evaluation: It processes items one-by-one only when requested.  
4. Multiple Iterables: You can pass more than one list to map(), provided your function accepts that many arguments.

## Q10 What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?  

- These three functions are known as Higher-Order Functions in Python.
- The easiest way to distinguish them is by their outcome:
  - filter(): Selects a subset (makes the list smaller).
  - map(): Transforms every item (changes the values).
  - reduce(): Combines everything into one (boils it down to a single value).  
1. filter():  
  - The filter() function creates a new list containing only the items that satisfy a specific condition (where the function returns True).  
  - Example:  
     ```  
       l = [2, 1, 3, 4, 4, 5, 6]
       list(filter(lambda x: x%2 ==0, l))
       #output - [2, 4, 4, 6] 
2. map(): 
  - It is a function which executes a specified function for each of iterm of an iterable.
  - It doesn't change the number of items, just their "shape" or value.  
  - syntax : map(func, *iterable)  
    ``` 
      l = [1, 2, 3, 4, 5]
      list(map(lambda x: x**2, l))
      #output - [1, 4, 9, 16, 25]
3. reduce():
  - It is also consider as folding / reduction of a function.
  - Reduce always take 2 argument function.  
  - It uses the result as the first argument for the next step until only one value remains.  
  - Example:  
     ```
      from functools import reduce
      l = [2, 1, 3, 4, 4, 5, 6]
      reduce(lambda x, y: x+y, l)
      #Output - 25

## Q1 Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.  

````python
def sum_even_numbers(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total
````

````python
sum_even_numbers([1,2,3,4,5,6])
````

````python
# compressed form
def sum_even_numbers(numbers):
    return sum([x for x in numbers if x % 2 == 0])
````

````python
sum_even_numbers([1, 2, 5, 7, 8])
````

````python
def sum_even_numbers(numbers):
    even_list = filter(lambda x: x % 2 == 0, numbers)
    return sum(even_list)
````

````python
sum_even_numbers([1,2,4,6])
````

## Q4 Write a Python function that checks if a given number is prime or not from 1 to 200.

````python
def prime_numb(num):
    if num < 2 or num > 200:
        return f"{num} is out of range"
    
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return f"{num} is not a prime number"
    return f"{num} is a prime number"
    
num = int(input("Enter the nubember between 1 to 200: "))
result = prime_numb(num)
print(result)
````

````python
# can be wriiten in another way

def is_prime(n):
    if n < 2 or n > 200:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
````

````python
is_prime(5)
````

## Q5  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

````python
class fibonaci_seq:
    def fib(n):
        a=0
        b=1
        for i in range(n):
            yield a
            a, b = b, a+b
gen = fibonaci_seq.fib(5)
gen
````

````python
next(gen)
````

````python
next(gen)
````

## Q6 Write a generator function in Python that yields the powers of 2 up to a given exponent.

````python
def power_2(n):
    for i in range(n + 1):
        yield 2**i
        
````

## Q7 Implement a generator function that reads a file line by line and yields each line as a string

````python
%%writefile test.txt
Apple
Banana
Cherry
Date
````

````python
def read_lines(filename):
    with open(filename, 'r') as f:
        yield from f

for line in read_lines("test.txt"):
    print(line.strip())
````
