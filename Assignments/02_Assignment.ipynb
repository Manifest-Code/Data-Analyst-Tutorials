## Q8 What is a hash table, and how does it relate to dictionaries in Python?  

- What is a hash table?  
  - A Hash Table is a data structure that stores organized data in a way that allows for nearly instantaneous retrieval.
  - In Python, the Dictionary (dict) is the high-level tool you use, but the Hash Table is the engine under the hood that makes it work.   

- How does it relate to dictionaries?  
    - When you store a key-value pair in a Python dictionary (e.g., {"name": "Alice"}), three main things happen:    
    1. Hashing: Python takes the key ("name") and runs it through a Hash Function. This function converts the text into a giant, unique  
       integer. 
    2. Compression: Since that integer is too big to be a memory index, Python "compresses" it (usually using a modulo operation) to fit  
       the size of its current internal table.  
    3. Storage: Python places the key and the value into a block at that specific index.    
    4. Speed (Complexity): Because the hash function tells Python exactly where to go, looking up a key takes the same amount of  
       time whether the dictionary has 10 items or 10 million.   
    5. Unique Keys: A hash table depends on keys being unique. If you try to add a second "name" key, it will hash to the same spot and  
       simply overwrite the old value.  
    6. Requirement for Immutability: For a hash table to work, a key's hash must never change. This is why you can use Tuples or Strings  
       as dictionary keys, but not Lists (because if the list changed, the hash would change, and the dictionary would "lose" the data).

## Q10 Explain why strings are immutable in Python?  

- String is a characters arranged in an unchangeable order.    
- Consider writing a sentence, a paragraph, or a piece of material.  
- In Python, once a string object is created, its content cannot be changed.   
- This isn't just a random design choice; it is a fundamental architectural decision that provides several critical benefits to the    
  language's performance and security.    
- It is immutable due to following reasons:  
  - Memory: To allow multiple variables to share the same string (Interning - When you create multiple string variables with the exact  
    same content, Python often points all of them to the same memory address rather than creating a new copy for each one.).  
  - Speed: To cache hash values for fast dictionary lookups.  
  - Safety: To prevent sensitive data (like URLs) from being changed mid-process.  
  - Logic: To avoid "side effects" where changing one variable affects others.

## Q13  How do sets handle duplicate values in Python?  

- In Python, sets handle duplicate values by automatically discarding them during initialization and update operations.  
- When you create a set or add an item to an existing set, Python checks if that value already exists. If it does, the new value is  
  simply ignored.    
- The "secret" to how sets handle duplicates so efficiently is Hashing.    
- Every time you add an item, Python runs it through a hash function to get a unique integer (a "fingerprint").    
- It uses this integer to determine exactly where that item should sit in the set's internal memory (the hash table).  
- If a new item generates the same hash and is found to be equal to an item already in that "block," Python realizes it's a duplicate  
  and doesn't add it.   


## Q14  How does the “in” keyword work differently for lists and dictionaries?  
 
- 1. The "in" Keyword with Lists:  
     - When you use item in my_list, Python performs a Linear Search. It starts at the very first index (0) and compares your item to  
       every single element one by one until it find a match or reaches the end.  
     - Logic: "Is the first item 'X'? No. Is the second item 'X'? No..."    
     - speed gets slower as the list grows larger.    

- 2. The "in" Keyword with Dictionaries:  
     - When you use key in my_dict, Python uses Hashing. It doesn't "search" at all; it calculates exactly where the key should be.  
     - Logic: It takes the key, runs it through a hash function to get a memory address, and jumps directly to that spot to see  
       if something is there.  
     - Speed is instant, regardless of size cause of hash function.  
      
 

## Q15 Can you modify the elements of a tuple? Explain why or why not?  

- No, you cannot modify the elements of a tuple once it has been created. In Python terminology, tuples are immutable.  
- Why can't you modify a tuple?  
    - Fixed Memory Allocation:   
      When a tuple is created, Python allocates a specific, fixed amount of memory to hold its contents. Because the size and contents are guaranteed not to change, Python doesn't need to add "extra space" for potential growth (unlike lists). This makes tuples more memory-efficient.  
    - Data Integrity and Safety:  
      The primary purpose of a tuple is to act as a constant record. By making them immutable. Data cannot be accidentally overwritten by other parts of your program.

## Q17  Describe the time complexity of accessing elements in a dictionary.  

- In Python, the time complexity for accessing an element in a dictionary by its key is O(1), also known as Constant Time.    
- The reason for this speed is the Hash Table architecture. Most data structures (like lists) require a "search," but a dictionary uses  
  a "calculation."    
- The Calculation: When you ask for my_dict["key"], Python doesn't look through the dictionary. Instead, it runs the key through a hash  
  function.      
- The Direct Jump: The hash function produces a specific integer that corresponds directly to a memory address.  
- The Retrieval: Python jumps straight to that memory address and grabs the value.  


## Q18  In what situations are lists preferred over dictionaries?  

- While dictionaries are superior for speed and labeling, Lists are the preferred choice when the sequence, order, and collection of  
  duplicates are more important than the lookup speed of a specific key.  
    
  1. When Order and Sequence Matter:
    - Lists are ordered by nature. They remember the exact sequence in which items were added. This makes them ideal for any data that  
       represents a timeline or a queue.  
    - Scenario: A playlist of songs, a history of undo actions, or a "To-Do" list.    
  2. When You Need to Store Duplicate Values:  
    - A dictionary cannot have duplicate keys. If you try to add a second entry for "Apple," it will simply overwrite the first one.    
      Lists, however, allow you to store the same value as many times as you want.    
    - Scenario: Recording the results of 100 coin flips.  
  3. When Performing Sequential Operations (Iteration):
    - If your primary task is to "do something to every item in this collection," a list is more memory-efficient and straightforward.  
    - Scenario: Applying a 10% discount to a list of prices.    
  4. Sorting: You need to rearrange items based on value (e.g., lowest to highest)  
      
  5. Slicing: You want to grab a specific range (e.g., my_list[2:5]).  
      
  6. Homogeneous Data: All items are the same type and don't need individual labels.  



## Q19 Why are dictionaries considered unordered, and how does that affect data retrieval?  

- Why are dictionaries considered unordered:  

  - Technically, dictionaries use a Hash Table.1 When you add an item, its position in memory is determined by a "hash function" (O(1)  
    logic), not by when you typed it.  
  - In older versions of Python (before 3.6), if you added {"a": 1, "b": 2, "c": 3}, the hash function might scatter them into memory  
    slots 5, 2, and 8. When you printed the dictionary, it would come out in a seemingly random order like {"b": 2, "a": 1, "c": 3}.  
  - Starting with Python 3.7, dictionaries are now officially ordered by insertion. This means Python remembers the sequence in which you  
    added keys.    
  - However, they are still considered "unordered" in a functional sense compared to lists:  
  - No Indexing: You cannot access the "first" item using my_dict[0]. You must use a Key.  
  - No Slicing: You cannot do my_dict[0:2] to get the first two entries.    
  
-  How does this affect data retrieval:  

  - Because the retrieval is based on Hashing rather than Position, the lack of "order" actually makes retrieval much faster.  
  - In a List (Ordered), to find a specific value, you often have to start at index 0 and look through every "ordered" slot until you  
    find it.  
  - In a Dictionary (Unordered/Hashed), Python ignores the "order" and jumps directly to the memory address.  
  - When you retrieve data from a dictionary, you are usually interacting with "Views" (.keys(), .values(), or .items()).  
  

## Q20  Explain the difference between a list and a dictionary in terms of data retrieval.  

- List:
  - In a list, elements are accessed via an integer index. To retrieve an item, you must know its exact location. If you don't know the  
    location and need to find a specific value, Python must look through the list one-by-one.  
  - Syntax: my_list[3]  
  - Search Style: Linear (scanning).    
  - Use a List for Retrieval when:    
    - The order of items is the most important factor.    
    - You need to retrieve items based on their sequence (e.g., "Give me the first three items").  
    - The data is a simple stream of values that you will always process from start to finish.  

- Dictionary:  
  - In a dictionary, elements are accessed via a unique key (like a string or integer). Python uses a "Hash Table" to map that key    
    directly to its value.    
  - Syntax: my_dict["price"]    
  - Search Style: Direct mapping (hashing).    
  - Use a Dictionary for Retrieval when:  
      - you have a unique identifier for your data (ID numbers, usernames, product codes).  
      - You need to look up information frequently in a large dataset.  
      - The relationship between the "Label" and the "Value" is more important than the order of the items.
