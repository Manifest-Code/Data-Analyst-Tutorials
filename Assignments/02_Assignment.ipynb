## Q.1 What are data structures, and why are they important?

- What are they?  
  Data structures are specialized formats for organizing, storing, and accessing collections of data. They provide efficient ways to  
  manage information based on its characteristics and intended use.  
  Think of them as containers that hold your data and determine how you can interact with it. Different containers are better suited for   different types of items.  

- Why are they important?  
  Choosing the right data structure significantly impacts the efficiency and performance of your program.  
  Well-chosen data structures can:  
  Simplify data manipulation (adding, removing, modifying elements)  
  Optimize searching and sorting operations  
  Conserve memory usage  

## Q.2 Explain the difference between mutable and immutable data types with examples

- Mutable DataTypes:    
  In mutable DataTypes you can change items even after they are created.   
  You can add, remove, or modify items without creating a completely new object in memory.  

  Examples: Lists, Dictionaries, Sets.  

  List = = [1, 2, 3]  
  List[0] = "Data"  
  -> ["Data", 2, 3]  

- Immutable data types:  
  In immutable DataTypes you can not change items once they are created.    
  If you try to modify an immutable object, Python actually creates a brand new object in a new memory location.  

  Examples: Strings, Tuples, Integers, Floats, Booleans.    

  t1 = (1, 2, 4)  
  t1[0] = "Data"  
  -> TypeError: 'tuple' object does not support item assignment

## Q8 What is a hash table, and how does it relate to dictionaries in Python?  

- What is a hash table?  
  - A Hash Table is a data structure that stores organized data in a way that allows for nearly instantaneous retrieval.
  - In Python, the Dictionary (dict) is the high-level tool you use, but the Hash Table is the engine under the hood that makes it work.   

- How does it relate to dictionaries?  
    - When you store a key-value pair in a Python dictionary (e.g., {"name": "Alice"}), three main things happen:    
    1. Hashing: Python takes the key ("name") and runs it through a Hash Function. This function converts the text into a giant, unique  
       integer. 
    2. Compression: Since that integer is too big to be a memory index, Python "compresses" it (usually using a modulo operation) to fit  
       the size of its current internal table.  
    3. Storage: Python places the key and the value into a "bucket" at that specific index.    
    4. Speed (O(1) Complexity): Because the hash function tells Python exactly where to go, looking up a key takes the same amount of  
       time whether the dictionary has 10 items or 10 million.   
    5. Unique Keys: A hash table depends on keys being unique. If you try to add a second "name" key, it will hash to the same spot and  
       simply overwrite the old value.  
    6. Requirement for Immutability: For a hash table to work, a key's hash must never change. This is why you can use Tuples or Strings  
       as dictionary keys, but not Lists (because if the list changed, the hash would change, and the dictionary would "lose" the data).    

  
      



## Q.9 Can lists contain different data types in Python?

- Yes, lists in Python are heterogeneous, meaning they can store a mix of different data types in a single collection.  
  Example of a Mixed-Type List  
  You can combine strings, integers, floats, booleans, and even other lists or dictionaries into one list:  

  list = [  
    "Python",           # String  
    2026,               # Integer  
    3.14,               # Float  
    True,               # Boolean  
    [1, 2, 3],          # Another List (Nested)  
    {"key": "value"}    # Dictionary  
]
