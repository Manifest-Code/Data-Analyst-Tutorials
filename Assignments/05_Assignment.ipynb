## Q1  What is the difference between interpreted and compiled languages?   

- Compiled Languages:
  - In this entire program is converted into machine code by a compiler
  - Then the machine code is executed directly by the computer.
  - Flow  
    Source Code → Compiler → Machine Code → Run
  - Examples    
     • C
     • C++
     • Java (compiled to bytecode)
     • Go
  - Features:  
     - Faster execution
     - Errors shown before running
     - No need source code at runtime
     - Platform dependent (need different executable for Windows/Linux)  

- Interpreted Languages:
  - Code is executed line by line by an interpreter
  - No separate compilation step.
  - Flow  
      Source Code → Interpreter → Execute line by line  
  - Examples  
      • Python
      • JavaScript
      • PHP
      • Ruby
  - Features
      - Easy to debug
      - Platform independent
      - No separate compilation
      - Slower than compiled languages


    Special Case – Python

    Python is actually:

    Interpreted + Compiled to bytecode (.pyc)

    Source → Bytecode → Python Virtual Machine
    That’s why Python is slower than C.

## Q11 What is memory management in Python?  

- Memory management in Python is the process of allocating, using, and freeing memory automatically while a Python program runs.
- Private Heap Space – all Python objects are stored here
- Python Memory Manager – controls allocation/deallocation
- Garbage Collector (GC) – removes unused objects  
- using Reference Counting:
   - Every object has a reference count
   - When a new reference is created    -> count increases
   - When reference is deleted          -> count decreases
   - When count becomes 0,              ->  memory is freed
   Example:
     a = [1,2,3]   # reference count = 1
     b = a         # reference count = 2
     del a         # reference count = 1
     del b         # reference count = 0  # memory released

## Q15  How does Python's garbage collection system work?  

- Python’s garbage collection (GC) system automatically removes objects that are no longer needed so that memory can be reused.
- It works mainly through:
   1. Reference Counting
   2. Garbage Collector for Cyclic References

- 1. Reference Counting (Main Method)
    - Every Python object has a reference count = number of variables pointing to it.
    - When the count becomes 0, the object is immediately deleted from memory.
    - Example:
      ```
          a = [1, 2, 3]   # reference count = 1
          b = a           # reference count = 2
          del a           # reference count = 1
          del b           # reference count = 0 → object deleted

- 2. Garbage Collector
     - Reference counting alone cannot handle this:
     - Example:
       ```
            list1 = []
            list2 = []

            list1.append(list2)
            list2.append(list1)

          Both objects refer to each other 
          i.e. reference count never becomes 0
     - To solve circular references, Python has a Garbage Collector module that:
     - Detects groups of objects referencing each other
     - Checks if they are unreachable from the program
     - Deletes them safely

## Q18  What is the difference between os.fork() and multiprocessing in Python?  

1. os.fork():
   - Low-level system call from the operating system
   - Available mainly on Unix/Linux (not Windows)
   - Creates an exact copy of the current process (child process)
   - Parent and child share the same memory initially  
   - Very fast
   - Direct OS-level process creation
   - Not portable (doesn’t work on Windows)
   - Hard to manage communication
   - No built-in tools like Queue, Pool  
   Example:  
    ```   
         import os

         pid = os.fork()

         if pid == 0:
            print("Child process")
         else:
            print("Parent process")
                  
2. multiprocessing Module:  
   - High-level Python module
   - Works on Windows, Linux, macOS
   - Provides tools like:  
      Process  
      Pool  
      Queue  
      Pipe  
      Shared memory  

   - Cross-platform
   - Easier and safer
   - Built-in communication
   - Better error handling
   - Slightly slower than fork
   - More memory usage

## Q23 What are the challenges associated with memory management in Python?

1. High Memory Consumption:
     - Python objects, especially lists, dictionaries, and classes, take more memory than similar structures in languages like C.
     - Example: Each integer in Python is an object with extra metadata.
     - Challenge: Large datasets can quickly consume a lot of memory.

2. Garbage Collection Overhead:
   - Python uses reference counting + garbage collection to free memory.
   - Objects are deleted when no references exist.
   - Problem: Circular references (objects referring to each other) can delay memory release until garbage collector runs.

        a = []
        b = [a]
        a.append(b)  # circular reference

      Even after del a and del b, memory may not be freed immediately.

3. Memory Leaks:
   - Memory leaks can occur if objects are kept alive unnecessarily, e.g., global lists, caches, or circular references.
   - Hard to detect because Python hides memory allocation details.

4. Large Object Allocation:
   - Python may struggle with very large objects (e.g., huge lists or arrays).
   - Copying or resizing large objects (like lists) creates temporary copies, using extra memory temporarily.

6. Inefficient Use of Memory for Immutable Objects
   - Immutable objects like strings and tuples are recreated instead of modified.
   - Frequent modification creates multiple temporary objects, increasing memory pressure.

7. Limited Control Over Memory
   - Unlike C/C++, Python developers cannot manually free memory.
   - You rely on Python’s garbage collector; sometimes you must manually delete objects or use gc.collect().

## Q25 Why is it important to use multithreading in certain applications?

- if you want single processor to use multiple thread to execute a process, it is called a multithrading.
- A program can divide big work into smaller parts (threads).
- These threads run “at the same time,” so the task finishes faster.
- Scince it happens on one processor, threads share same memory space and resources (data, variables, executable program) within a process.  
- concurrent  
- use case: Best for tasks that are part of same applicatopm and need to share data or resources.
Example:
Downloading 5 files → each file can download in a separate thread instead of one-by-one.  

``` 
        import time
        import concurrent.futures
        start = time.perf_counter()

        url_list = [
            'https://raw.githubusercontent.com/dscape/spell/master/test/resources/big.txt',
            'https://raw.githubusercontent.com/first20hours/google-10000-english/master/google-10000-english-no-swears.txt',
            'https://raw.githubusercontent.com/itsfoss/text-files/master/sherlock.txt' ,
            'https://raw.githubusercontent.com/itsfoss/text-files/master/sample_log_file.txt',
        ]


        data_list = ['data1.txt', 'data2.txt', 'data3.txt', 'data4.txt']
            
        import urllib.request
        def file_download(url, filename):
            urllib.request.urlretrieve(url, filename)

        with concurrent.futures.ThreadPoolExecutor() as executor:
            executor.map(file_download, url_list, data_list)



        end = time.perf_counter()


        print(f"The program finished in {round(end-start, 2)} seconds.")

````python
import logging

logging.basicConfig(filename = "division_error.log", level = logging.INFO, format = '%(asctime)s %(levelname)s %(message)s')

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b
    logging.info(f"Division successful: {a} / {b} = {result}")
    print("Result:", result)

except ZeroDivisionError:
    logging.error("Division by zero attempted")

except ValueError:
    logging.error("Invalid input entered")

logging.shutdown()
````

## Q15  Write a Python program that prints the content of a file and handles the case when the file is empty?

````python
try:
    with open("example_text.txt", "r") as file:
        content = file.read()

        if content.strip() == "":
            print("The file is empty.")
        else:
            print(content)

except FileNotFoundError:
    print("Error: File does not exist.")
````

````python
# Another way 

try:
    with open("example_text.txt", "r") as file:
        lines = file.readlines()

        if not lines:
            print("The file is empty.")
        else:
            for line in lines:
                print(line, end="")

except FileNotFoundError:
    print("File not found.")
````

## Q18 How would you implement a basic logging setup that logs to a file with rotation after 1MB?

````python
import logging
from logging.handlers import RotatingFileHandler

# 1. Setup the handler
# maxBytes=10**6 is approximately 1MB
# backupCount=3 keeps the current log + 3 old versions
handler = RotatingFileHandler("logfile.log", maxBytes=1_000_000, backupCount=3)

# 2. Configure the logging format
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s", handlers=[handler])

# 3. Test it
logging.info("This is a test log entry.")
````

## Q21 Write a Python program that reads a file and prints the number of occurrences of a specific word

````python
def count_word_in_file(filename, word):
    with open(filename, 'r') as file:
        content = file.read()
        content = content.lower()
        print(content)
        count = content.count(word.lower())
        print("The count of", word, "is:", count)
count_word_in_file("test.txt", "Cherry")
````

## Q22  How can you check if a file is empty before attempting to read its contents?

````python
import os
os.path.getsize("empty_file.txt") # get size of file
````

````python
def is_empty(filename):
    with open(filename, 'r') as f:
        content = f.read(1).strip()  # read first character and strip whitespace
        return not content

if is_empty("empty_file.txt"):
    print("The file is empty.")
````

## Q23  Write a Python program that writes to a log file when an error occurs during file handling.

````python
import logging
from os import path

logging.basicConfig(level = logging.DEBUG, format = '%(asctime)s %(levelname)s %(message)s')

def error_handling_file_read(file_path):
    try:
        with open(file_path, "r") as f:
            return f.read()
        logging.info("File read successfully.")
    
    except FileNotFoundError as e:
        logging.error(f"File Not Found Error occurred: {e}")
    except PermissionError as e:
        logging.error(f"Permission Error occurred: {e}")
    except Exception as e:
        logging.error(f"An unexpected error occurred: {e}")
        
error_handling_file_read("file.log")   
````
