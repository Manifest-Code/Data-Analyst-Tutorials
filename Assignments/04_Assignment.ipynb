## Q6  Explain the concept of inheritance in OOP?  

- Inheritance is a mechanism that allows a new class (the subclass or child) to acquire the properties (attributes) and behaviors (methods) of  
  an existing class (the superclass or parent).  
- Inheritance is Used for code reusability and ease of maintenance.
- Types of Inheritance:  
   - Single Inheritance: A subclass inherits from exactly one parent class (e.g., Dog inherits from Animal).
   - Multilevel Inheritance: A class inherits from a subclass, creating a chain (e.g., Puppy inherits from Dog, which inherits from Animal).
   - Hierarchical Inheritance: Multiple subclasses inherit from one single parent class (e.g., both Car and Truck inherit from Vehicle).
   - Multiple Inheritance: A class inherits from more than one parent class. While supported in languages like Python and C++, Java avoids it for  
     classes to prevent ambiguity (often called the "Diamond Problem").
   - Hybrid Inheritance: A combination of two or more of the types above.

## Q8 How is encapsulation achieved in Python?  

- Encapsulation is achieved through naming conventions and property decorators.  
  1. Naming Conventions (Access Levels): 
     Python uses underscores to signal intended access levels to other developers.  
       - Public (Default): Attributes without underscores are accessible from anywhere. Example: self.name.  
       - Protected (_): Prefixed with a single underscore (e.g., self._balance). This is a convention indicating the variable is for internal use   
         within the class and its subclasses; it is still technically accessible but should not be touched directly from outside.  
       - Private (__): Prefixed with double underscores (e.g., self.__salary). This triggers Name Mangling, where Python internally renames the  
         attribute to _ClassName__attribute to prevent accidental external access. 
  2. The @property Decorator (Getters and Setters):    
     The way to handle encapsulation is by using the @property decorator to create managed attributes. This allows you to access a method like a   regular attribute while adding logic for validation or security behind the scenes.  
       - Getter: Use @property to define a method that retrieves a private value.  
       - Setter: Use @attribute_name.setter to define a method that validates data before it is saved.  
       - Deleter: Use @attribute_name.deleter to control how an attribute is deleted.  
       
       
     Example:  
      
      ```
        
         class Student:
           def __init__(self, name, price):
              self.__name = name
              self.__price = price

           @property
           def access_price(self):
              return self.__price

           @access_price.setter    
           def set_price(self, new_price):
              self.__price = new_price

           @access_price.deleter
           def access_price(self):
              del self.__price  

         stud = Student("Alice", 1000)  
         stud.access_price                 #output - 1000

         # Accesses the setter
         stud.set_price = 1500 
         stud.access_price                 #output - 1500      
         
         # Raises AttributeError
         del stud.access_price 
         stud.access_price                 #output - AttributeError: 'Student' object has no attribute '_Student__price'

         


            

## Q9 What is a constructor in Python?  

- A constructor is a special method used to create and initialize a new object from a class.  
- The Initializer: __init__
    - The __init__ method is automatically called immediately after a new object is created. Its primary purpose is to assign initial values to the  
      object's attributes. 
    - Naming Convention: It must be named exactly __init__ (with double underscores).
    - The self Parameter: The first argument must always be self, which represents the specific instance being created.
    - Return Value: It must always return None. Attempting to return any other value will raise a TypeError.  
- Types of Constructors:  
    - Parameterized Constructor: Accepts arguments (besides self) to allow creating objects with unique initial values.
    - Non-Parameterized Constructor: Only takes self and initializes every object with the same default values.
    - Default Constructor: If you do not define an __init__ method, Python automatically provides an empty one that does nothing but create a plain  
      object.    
    Example:  
         ``` 
         class Student:
            def __init__(self, name, degree):
                self.name = name                      # Initialize attribute                
                self.degree = degree                  # Initialize attribute
    
            def show(self): 
                print("name", self.name, "degree", self.degree)  
            
            # Creating objects (instantiation)
            a = Student("Alice", "Masters")  
            a.show()                                 #output - name Alice degree Masters

         

## Q10 What are class and static methods in Python?  

1. Class Methods (@classmethod):
   - A class method is bound to the class itself rather than any single instance of it. 
   - Decorator: Defined using the @classmethod decorator.
   - Implicit Argument: It automatically receives the class as its first argument, conventionally named cls.
   - Primary Use: Best suited for factory methods—alternative constructors that create class instances in specific ways (e.g., creating a User from  
     a JSON string).
   - State Access: It can access and modify the class state (class-level variables) which affects all instances, but it cannot access  
     instance-specific data (self).  

2. Static Methods (@staticmethod):  
   - A static method is essentially a regular function that lives inside a class's namespace for organizational purposes. 
   - Decorator: Defined using the @staticmethod decorator.
   - Implicit Argument: It receives no implicit arguments (neither self nor cls).
   - Primary Use: Used for utility or helper functions that perform a task related to the class but do not need to read or change its data (e.g., a  
     math conversion or a validation check).
   - State Access: It cannot access or modify either the class state or the instance state.

## Q12 What is method overriding in OOP?  

- Method overriding is a language feature that allows a subclass (child class) to provide its own specific implementation of a method that is  
  already defined in its superclass (parent class).  
- The overriding method must have the exact same name and parameters as the method in the parent class.  
- Overriding only occurs between a parent and child class; it cannot happen within a single class.  
- The access modifier for the overriding method must be the same as or less restrictive than the one in the superclass (e.g., a protected method  
  can be overridden as public, but not as private).    

   Example:   
    ```
          class Animal:
              def sound(self):
                  print("Animal Sound")
          class Cat(Animal):
              def sound(self):
                  print("Cat Sound")  

          anm = Animal()
          anm.sound()                   # output - Animal Sound

          cat = Cat()
          cat.sound()                   # output - Cat Sound 

## Q13  What is a property decorator in Python?  

- The @property decorator is a built-in tool that allows you to turn a class method into an "attribute".  
- The property decorator allows you to define three distinct behaviors for a single attribute name: 
   1. Getter (@property): Triggered when you read the value (e.g., obj.name).
   2. Setter (@name.setter): Triggered when you assign a value (e.g., obj.name = "Alice").
   3. Deleter (@name.deleter): Triggered when you delete the attribute (e.g., del obj.name).  

   Example: 
    ``` 
       class Student:
           def __init__(self, name, price):
               self.__name = name
               self.__price = price

           @property
           def access_price(self):
               return self.__price  

           @access_price.setter    
           def set_price(self, new_price):
               self.__price = new_price

           @access_price.deleter
           def access_price(self):
               del self.__price


       stud = Student("Alice", 1000)  
       stud.access_price               # output - 1000  

       
       stud.set_price = 1500
       stud.access_price               # output - 1500

       del stud.access_price

## Q17 What is the difference between a class variable and an instance variable?  

1. Class Variables:
   - Class variables are shared across all instances of a class. They are defined directly inside the class body but outside any methods.
   - Ownership: Belong to the class itself.
   - Behavior: If the class variable is updated, the change is reflected in every object of that class.
   - Use Case: Storing constants, default settings, or counters (e.g., tracking the total number of employees).  

2. Instance Variables:
   - Instance variables are unique to each object. They are defined inside the constructor method (__init__) and are prefixed with self.
   - Ownership: Belong to the specific object.
   - Behavior: Changing the value for one object does not affect others.
   - Use Case: Storing unique data like a user's name or a product's ID.


  Example:  

  ```  
      class Student:
    
          total_students = 0                    #class variable
          def __init__(self, name):             #instance method
              self.name = name                  #instance variable
              Student.total_students = Student.total_students+1 #total_student is the class variable associated with all the instances
        
          @classmethod
          def get_total_students(cls):
              return cls.total_students   

          Student.total_students       # output - 0

          std1 = Student("Alice")
          std2 = Student("Ben")

          std1.name                    # output - 'Alice'
          std2.name                    # output - 'Ben'
          Student.total_students       # output - 2


## Q18 What is multiple inheritance in Python?  

- A feature that allows a single class (the child or derived class) to inherit attributes and methods from more than one parent class (superclasses).  
- It combines features from different classes without rewriting code.  
- Easily models complex real-world relationships.  
- Syntax: class Child(Parent1, Parent2):  
- Method Resolution Order (MRO): When a method is called, Python searches for it in a specific order: first in the child class, then in the  
  parent classes from left to right as they were listed in the class definition.
- super() Function: This function is used to call methods from the parent hierarchy.  
- A classic challenge in multiple inheritance is the Diamond Problem, which occurs when two parent classes inherit from the same common  
  grandparent class, and a child class inherits from both. Python resolves this ambiguity using the C3 Linearization algorithm, which ensures each class in the hierarchy is visited exactly once.  

  Example:  
  ``` 
     class Parent1:
         def parent1_info(self):                # method is same in both parents
             print("Inside parent1 class")
     class Parent2:
         def parent1_info(self):                # same method
             print("Inside parent2 class")
     class ChildClass(Parent1, Parent2):
         def child_info(self):               
             print("Inside child class")

     c2 = ChildClass()
     c2.parent1_info()  #the parent class which is inherited first in child class, that method will be executed. in this case parent 1  

     # output
     Inside parent1 class

1. __str__ (The Informal Representation)
  - Purpose: To provide a human-readable, user-friendly description of an object.
  - Audience: End users of the program.
  - Triggered By: The print() function, str() built-in, or f-string formatting ({obj!s}).

2. __repr__ (The Official Representation)
  - Purpose: To provide a detailed, unambiguous, and precise description of an object.
  - Audience: Developers and programmers for debugging and maintenance.
  - Triggered By: The repr() function, the Python REPL (interactive shell), or debuggers.
  - Goal: Ideally, the string returned should be a valid Python expression that can be used to recreate the object (e.g., ClassName(attr1=val1)).

## Q2 Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle from it and implement the area() method in both.

````python
import abc
class Shape:
    @abc.abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        radius = self.radius
        return 3.14 * radius**2
    
class Rectangle(Shape):
    def __init__(self, lenght, breadth):
        self.len = lenght
        self.brth = breadth
    def area(self):
        lenght = self.len
        breadth = self.brth
        return lenght * breadth
````

````python
cir = Circle(5)
````

````python
cir.radius
````

````python
cir.area()
````

````python
rect = Rectangle(3, 4)
````

````python
rect.len
````

````python
rect.area()
````

## Q5 Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes balance and methods to deposit, withdraw, and check balance.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    def deposit(self, amount):
        self.__balance += amount
    def withdraw(self, with_amount):
        if self.__balance >= with_amount:
            self.__balance -= with_amount
            return (f"Withdrawn: {with_amount}")
        else:
            return "There is no efficienct balance in your account to withdraw this amount"
    def check_balance(self):
        return self.__balance
````

## Q14 Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary.

````python
class Employee:
    def __init__(self, hours_worked, hourly_rate):
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate
    
    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate
    
class Manager(Employee):
    def __init__(self, hours_worked, hourly_rate, bonus):
        super().__init__(hours_worked, hourly_rate)
        self.bonus = bonus

    def salary_with_bonus(self):
        return super().calculate_salary() + self.bonus
        
````

````python
emp1 = Employee(5, 700)
````

````python
emp1.calculate_salary()
````

````python
emp1 = Manager(5,700, 500)
````

````python
emp1.salary_with_bonus()
````
