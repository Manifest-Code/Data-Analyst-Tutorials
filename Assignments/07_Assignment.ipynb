## Q1. What is a RESTful API?  

- RESTful API is a way for two computers to talk to each other over the internet using the HTTP protocol.  
- A RESTful API relies on standard HTTP methods to tell the server exactly what you want to do with a piece of data.
- HTTP Methods:  
   - **POST:** Creates a new resource   
   - **GET:** Retrieves a resource or collection of resources   
   - **PUT:** Updates an existing resource   
   - **DELETE:** Remove a resource   
      
  ## Principles of REST API  

- Statelessness: The server doesn't remember you from one request to the next. Every single request must contain all the information needed to understand it (like providing your ID token every time).

- Client-Server Separation: The front-end (what the user sees) and the back-end (where the data lives) are totally separate. You can change the website's look without breaking the API.

- Uniform Interface: Resources are always identified by URLs.

   Example: http://127.0.0.1:5000/get-user/123 always refers to user 123.

- JSON Format: While not strictly required by the REST "rules," almost all modern REST APIs exchange data using JSON because it’s lightweight and easy for both humans and machines to read.

## Q2. What is Flask, and why is it popular for building APIs?  

- What is Flask?
  - Flask is a micro-framework written in Python.  
  - Unlike "batteries-included" frameworks like Django (which comes with an admin panel, database engine, and user authentication built-in), Flask is minimal.
  - It handles Routing (matching a URL to a Python function).  
  - It handles Requests/Responses (getting data from a user and sending it back).  
  - Everything else (like the Caching you used earlier) is added only when you need it.  
- Why is it so popular for APIs?  
  - 1. Simplicity and Speed: You can write a fully functional API in about 5 lines of code. There is very little "boilerplate" code, meaning you spend more time writing your logic and less time configuring the framework.
  - 2. The "Data Science" Connection: Since most Machine Learning models are built using Python (Scikit-Learn, TensorFlow, PyTorch), Flask is the most natural way to "wrap" those models into a web service. You don't have to switch languages to put your model online.
  - 3. Flexibility: Because it’s a micro-framework, you aren't locked into a specific way of doing things. you can use Redis for caching, or can use SQLAlchemy for your database also if you Want to keep it tiny with no database at all you can do that.

## Q4 What is the purpose of the @app.route() decorator in Flask?  

- The @app.route() decorator is the GPS of your Flask application.  
- Its primary purpose is to bind a specific URL (the web address) to a specific Python function (the logic).
- When a user types a URL into their browser or Postman, Flask looks at its "map" of decorators to figure out which function it should run to give the user an answer.  
- Example:  
  ```
    @app.route('/hello')
    def hello():  
      return "Hello, World!"  

- The Trigger: A user visits yourdomain.com/hello.
- The Match: Flask sees the @app.route('/hello') decorator.
- The Action: Flask automatically calls the hello() function.
- The Result: The string "Hello, World!" is sent back to the user's browser.

## Q5 What is the role of Flask-SQLAlchemy?  

- Think of Flask-SQLAlchemy as a "translator" between your Python code and your database.
- In a normal database, you have to write SQL (Structured Query Language) to get things done. With Flask-SQLAlchemy, you can treat your database tables like regular Python Classes and your data rows like Python Objects
- This concept is called ORM (Object-Relational Mapping).   

- Example:  

      ```
      from flask_sqlalchemy import SQLAlchemy

      db = SQLAlchemy(app)

      # This class 'User' creates a table in your database automatically
      class User(db.Model):
            id = db.Column(db.Integer, primary_key=True)
            username = db.Column(db.String(80), unique=True, nullable=False)
            email = db.Column(db.String(120), unique=True, nullable=False)

      # To add a new person to your database:
      new_user = User(username="JohnDoe", email="john@example.com")
      db.session.add(new_user)
      db.session.commit()

## Q6 How do you create a basic Flask application?  

1. Import flask:
   - **from flask import Flask** pulls in the core class. This class is the foundation of your entire web application.  
2. Create instance:  
   - app = Flask(__name__) creates the "app object."
   - The __name__ argument is a special Python variable that tells Flask where to look for resources like templates and static files (images/CSS).  
3. Define function:  
   - def home(): is a standard Python function. Whatever this function returns is what the user will see in their browser (usually HTML, text, or JSON).  
4. Run the application:  
   - app.run(debug=True) starts the local development server.
   - debug=True automatically restarts the server every time you save a change to your code, and it provides helpful error messages in the browser if something crashes.  

- Example:
   ```
   # import falsk
   from flask import Flask

   # 1. Initialize the app
   app = Flask(__name__)

   # 2. Define a function
   @app.route('/')
   def home():
      return "Hello, Flask is running!"

   # 3. Run the application
   if __name__ == "__main__":
      app.run(debug=True)

## Q7 How do you return JSON responses in Flask?  

1. The Automatic Way:  
   - If your function returns a Python dictionary or a list, Flask automatically calls jsonify() behind the scenes and sets the Content-Type to application/json.  
   - Example:  
    ```
    @app.route('/user/data')
    def get_data():
        # Flask sees this dict and converts it to JSON automatically
        return {"status": "success", "value": 42}  
2. Using jsonify() (The Explicit Way):  
  - While the automatic way is great, using the jsonify() function is better if you need to return lists or want to be more explicit about your intentions.  
  - Example:  
    ```
    from flask import jsonify

    @app.route('/user/predict')
    def predict():
        data = [1.2, 3.4, 5.6]
        return jsonify(results=data)   

3. Returning JSON with HTTP Status Codes:  
  - When building an API, you often need to tell the client if something went wrong (like a 404 Not Found) or if a resource was created (201 Created). You do this by returning a tuple:  
  - Example:  
    ```
    @app.route('/user/error')
    def error_example():  
       return jsonify({"error": "Resource not found"}), 404

## Q9 How do you handle errors in Flask (e.g., 404)?  

-  In Flask, handling errors like a 404 (Not Found) or a 500 (Internal Server Error) is done using the @app.errorhandler() decorator.  
- 1. Catching a 404 Error (Page Not Found)
   - This is the most common error. It happens when a user types a URL that doesn't exist in your @app.route list.  
   - Example:  
     ```
     @app.errorhandler(404)
     def handle_404(e):
        return jsonify({"error": "Resource not found", "status": 404}), 404   
- 2. Catching a 500 Error (Server Crash):  
   - This is what you were seeing earlier! A 500 error happens when your Python code crashes (like a FileNotFoundError or TypeError).
   - Example:  
     ```
      @app.errorhandler(500)
      def internal_server_error(e):
         return jsonify({
         "error": "The server encountered a problem",
         "details": str(e) # Useful for debugging, but hide this in production!
         }), 500


## Q10  How do you structure a Flask app using Blueprints?  

- As your Flask application grows, putting every route in a single app.py file becomes a nightmare to manage. 
- Blueprints are Flask's way of organizing your app into smaller, independent modules (like "folders" for your features).  
- Think of Blueprint as a mini-app that you "register" into your main application later.  
- For a professional API, you should separate your logic like this:  
  ```
    my_project/
    ├── app.py              # The "Manager" (Initializes everything)
    ├── routes/
    │   ├── __init__.py     # Makes 'routes' a Python package
    │   ├── predict.py      # Blueprint for ML predictions
    │   └── auth.py         # Blueprint for user login
    └── models.py           # Database models  

- 2. Creating a Blueprint (routes/predict.py)
In your specific feature file, you define a Blueprint object instead of a Flask app object.
 - Example:  
   ```
   from flask import Blueprint, jsonify
   
   # Create the Blueprint: (name, import_name, url_prefix)
   predict_bp = Blueprint("predict",__name__)

   @predict_bp.route("/predict", methods = ['POST'])
   def get_prediction():  
      return jsonify({"message": "Prediction successful!"})  

- 3. Registering the Blueprint (app.py):  
   - Now, you tell your main Flask app to "include" those routes.  
   - Example:  
     ```
     from flask import Flask, request
     from routes import predict, user

    app = Flask(__name__)

    ## import your blueprints here and register
    app.register_blueprint(predict.predict_bp)

    if __name__ == '__main__':
        app.run(debug=True)
