## access rows and columns of a dataframe >>  
- implicit/integer/internal indexing. - Normal indexing consists integer for indexing.
- explicit or named indexing. - consists of indexing as a name 

```python
#iloc
df.iloc[0:3] #start from 0 and go till 2. gives integer indexes
```

```python
#iloc
df.loc[0:3] #start from 0 and go till 3. gives name indexes. it cosiders 0 as name of a row, 1 as a name of a row
```

```python
df.iloc[0:2, ["Name", "Sex"]] #it will throw an error as iloc only shows integer indexes and here are name indexes name and sex
```

```python
df.loc[0:2, ["Name", "Sex"]] # instead use loc it won't throw erroe
```

```python
df.iloc[0:2,3:6]  # 0 to 1 rows and 3 to 5 columns (0-indexed)
```

## Remove the column from original dataframe

```python
df.drop('PassengerId') #need to provide along which axis want to drop. axis = 0 means drop from rows, axis = 1 means drop from columns
```

```python
df.drop('PassengerId', axis=1) #axis = 0 means drop from rows, axis = 1 means drop from columns, but it will not drop from original dataframe
```

```python
df # Passanger id is still in original dataframe
```

```python
df.drop('PassengerId', axis=1, inplace=True) #axis = 0 means drop from rows, axis = 1 means drop from columns, inplace=True means modify the original dataframe
```
