## Merge/ join 2 dataframes  
1. Inner Merge: If you are merging base on common element is called inner merge(intersection)  
2. Left Merge: Whaterever there is on dataframe A base on that you are merging is called left merge  
3. Right Merge: Based on All elements on right dataframe B are merge is called right merge 
4. Outer Merge: all the common element of the common column from both the dataframe will be printed.
5. Cross Merge: complete merge of all the dataframes. outer merge. is called cross merge.    Every permutation of rows are there 

```python
#merge
df1 = pd.DataFrame({'key1':[1,2,4,5,6],
                      'key2':[4,5,6,7,8],
                      'key3':[3,4,5,6,6]
})
df1
```

```python
df2 =  pd.DataFrame({'key1':[1,2,45,6,67],
                      'key4':[56,5,6,7,8],
                      'key5':[3,56,5,6,6]
}
)
df2
```

```python
pd.merge(df1, df2, how = "inner") # here key1 is common column in both and 1, 2, 6 values of key 1 column is same
```

```python
pd.merge(df1, df2, how = 'left') # all the values of df1 is printed and common values are 1, 2, 6 are there and rest are NA
```

```python
pd.merge(df1, df2, how = 'right') # all the values of df2 is printed and common values are 1, 2, 6 are there and rest are NA
```

```python
pd.merge(df1, df2, how = "outer") #all the common element of the common column will be printed and common values are 1, 2, 6 are there and rest are NA 
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1, df2, how = "cross") # each row of df1 will be repeated for all rows of df2. i.e. row 1 of df2 is repeated 5 times for 5 rows of df2
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1, df2, how = 'left', left_on="key2", right_on="key4") # in this takes common values from key2 from df1 and key 4 from df2. i.e. 5,6,7,8 are common in both key2 and key4
# and science key1 we are taking common even though it is, so another key1 for df2 is created as its not common
```

## join >> join works based on index/merge works based on columns

```python
df1 = pd.DataFrame({'key1':[1,2,4,5,6],
                      'key2':[4,5,6,7,8],
                      'key3':[3,4,5,6,6]},
                    index = ['a', 'b', 'c', 'd', 'e']
)
```

```python
df1
```

```python
df2 =  pd.DataFrame({'key6':[1,2,45,6,67],
                      'key4':[56,5,6,7,8],
                      'key5':[3,56,5,6,6]
},
index = ['a', 'b', 'h', 'i', 'j']
)
df2
```

```python
df1.join(df2, how = "inner") #joins columns of another dataframe based on common index
# here a and b index are common in both df
```

```python
df1.join(df2, how = "left") #all elements of df1 printed and common index of values from both dfs i.e. a, b remainng are NA
```

```python
df1.join(df2, how = "right") # all elements of df2 printed and common index of values from both dfs i.e. a, b remainng are NA
```

```python
df1.join(df2, how = "outer") # printed coomoun indexes and common values, remaining are NA
```

```python
df1.join(df2, how = "cross") # each row of df1 will be repeated 5 times for 5 rows of df2
```
