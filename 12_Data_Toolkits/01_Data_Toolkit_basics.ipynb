## Process CRISP-DM Framework

Cross-Industry Standard Process for Data Mining - Getting insights from data.  

1. Business Understanding: Problem Statement: To predict the price of a house. Area of house No. of rooms  
2. Data Understanding: How the area of a house decides the price of a house. Area of House - Price of House - No. of rooms - Price - No. of rooms - Area - Exploratory Data Analysis (EDA)  
3. Data Preparation: Formatting and cleaning data for use.
4. Modelling: Building mathematical representations.
5. Evalution: Assessing model performance.
6. Deployment: Implementing the solution into a real-world environment.

## Libraries, Packages, and Modules, functions(consist of code which can be reuse again)  

- Functions (e.g., Addition, Subtraction) -> grouped into Modules.
- Modules (e.g., Basic Calculations, Linear Algebra, Calculus) -> kept together as a Package.
- Multiple Packages -> called a Library.
