```python
#!pip install bokeh
import bokeh.io
import bokeh.plotting
bokeh.io.output_notebook()
```

```python
from bokeh.sampledata.iris import flowers
from bokeh.plotting import figure, output_file, show
```

```python
flowers
```

```python

output_file("test.html")

p = figure(title = 'test_flower')
p.xaxis.axis_label = "petal_length"
p.yaxis.axis_label = "petal_width"

#  has a set size (avoids unset radius)
p.scatter(flowers['petal_length'], flowers['petal_width'], size=10, marker="circle")

show(p)   # it also gives html file for the
```

```python
output_file("test.html")
p = figure(title = 'test_flower')
p.xaxis.axis_label = "petal_length"
p.yaxis.axis_label = "petal_width"
p.line(flowers['petal_length'], flowers['petal_width']) #line for line chart
show(p)
```

```python
x = [1, 2, 3, 4]
y = [3, 8, 4, 7]

p = figure(title="line_chart")
p.line(x, y)
show(p)
```

```python
x = [1, 2, 3, 4]
y = [3, 8, 4, 7]

p = figure(title="line_chart")
p.scatter(x, y, size = 20, fill_color = 'red', legend_label = "This is my Data")
show(p)
```
