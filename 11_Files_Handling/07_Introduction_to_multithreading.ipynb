- code -> program - > process    

- process is a instance of a program  

- example:

  - and you run a function test() - will take 1 sec  
  - run another function   test() - will take 1 sec supposly  
  -   so the totla time taken will be  2 sec  
  - but in computer there are numbers of processors like 11, 10 so we can use them and here comes the concept of running code parallely.

  - so multiple processors  are used to run the programm so its called **multi Processing**  

## Multi Threading  
 
-  program -> process -> divided into mutiple threads  
- Thread is the smallest unit of a process/processing  
- i.e. process -> processor 1 -> have divided into multiple threads.

- by defualt a process in a process runs on a single thread is called single thread process.

- But if you want to use multiple thread to execute a process, it is called a multithrading.  

- i.e one processer -> multiple thread -> **multiple threading**

- multi threading executes concerrently  


## Concerrency  

- test1() -> takes 10 sec to execute but after only 5 sec you run test2 which also takes 10 sec.  
- so at one point of time only one thread will be executing this process called **concerrancy**  
- This happens cause of GIL (global Interpreter lock) in pyrhon which says only one thread can run at a time.

## Difference between Multi Threading and Multi Processing
- Multi Threading:  
  - Scince it happens on one processor, threads share same memory space and resources (data, variables, executable program) within a process.  
  - concurrent  
  - use case: Best for tasks that are part of same applicatopm and need to share data or resources.  
   example:  
   downloding files, managing tabs on browser


- Multi Processing:  
  -  Multiple core/processors are there, each process has its own memory space and resources which are isolated from other process.  
  - Turely parallel on mutiple CPU cores (cause no GIL)   
  - Use case: suitable for tasks that are independent and can run in isolation  
  Example:  
    some computation, server handling, multiple request simultaniously
