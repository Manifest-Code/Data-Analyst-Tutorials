## Why need to create a different scripts?  
- To maintain the code, to improve understanding of a code
- Best practice is to keep different componenets in different scripts  
- Example: model training and EDA(Exploratory Data Analysis) should be kept in differnet script  

## Module:  
- A module is a single file (with .py extention) that contains python code  
- inside module fuctions, class or variables or some python code is present  


## Packages:  
- A package is a collection of modules arrange in a directories  
- each directory can have multiple python scripts  
- before python versoin 3.3 to make packages it was necessary to include __init__.py in package (to initialize pakage and expose required modules and functions) i.e file was needed to created with __init__.py name if you see it in some older projects in company you can use it to initialize the function  - not required anymore

## Library:  
- It is a collection of multiple modules and packages.  
- It is a pre- written code to perform common task.  
example: numpy, Pandas, matplotlib
