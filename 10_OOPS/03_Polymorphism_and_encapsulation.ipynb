```python
class Teacher_lecture:
    def lec_info(self):
        print("This is lec info with teacher perspective")
class Student_lecture:
    def lec_info(self):
        print("This is lec info with student perspective")
```

```python
obj1 = Teacher_lecture()
obj2 = Student_lecture()
class_obj = [obj1, obj2]
```

```python
def Parcent(class_obj):
    for i in class_obj:
        i.lec_info()        # at this point, lec_info is taking two forms wrt teacher and student
```

```python
Parcent(class_obj)
```

## 🌷 Method Overloading ✨  

- Polymorphism is oops takes place in a two ways:  
- Method Overloading: - python doesn't support true method overloading (unlike in java, in python you can achieve overloading but its not true overloading)  
- Method overrriding:  
- It always happens in the same class. 

## 💎 Method Overloading

```python
class Student:
    def student(self):
        print("Welcome to DataAnalyst Tutorial")
    def student(self, name = ""):
        print("Welcome to DataAnalyst Tutorial", name)
    def student(self, name = "", course = ""):
        print("Welcome to DataAnalyst Tutorial", name, course)
```

```python
stud = Student()
```

```python
stud.student()
```

```python
stud.student("Alice")    # if you pass argument 2nd method will be executed otherwise it will execute 1st method
```

```python
stud.student("Alice", "Ds")  # passed 2 argument 3rd method will be executed
```

```python
# this method alone can do all the above operations - this is called method overloading
class Student:
    def student(self, name = "", course = ""):
        print("Welcome to DataAnalyst Tutorial", name, course)

# student method is taking differnet forms, the last method overloads the previous once in the same class
```

```python
stud = Student()
stud.student()
```

```python
stud.student("Alice")
```

```python
stud.student("Alice", "DS")
```

## 🍁 Encapculation 🪄  

- Encapculation means hiding something
- bundling of data and methods of a class
- access modifer - public, protected, private


```python
class Student:
    def __init__(self, name, degree):
        self.name = name                   # this is public access modifier can be use outside of class as well
        self.degree = degree
```

```python
stud1 = Student("Ben", "Masters")
```

```python
stud1.name
```

```python
stud1.degree
```

```python
stud2 = Student("Sam", "Batch")
stud2.degree
```

```python
stud2.degree = "Phd"
stud2.degree                   # this is working cause data is public
```

- Public Access Modifier are accessible from anywhere from outside of class

- Private Modifier - the data and method is only accessible within its class, use __ (double underscore) to make data member private  

- but if you want to access this private method can access it by calling  _class__degree (single underscore class and double underscore method)

- and you can call private atribute by defining it in another function. Also the private method you can call it the same way.


```python
class Student:
    def __init__(self, name, degree):
        self.name = name                   
        self.degree = degree
    
    def show(self):
        #accessing the private and public data 
        print("name", self.name, "degree", self.degree)
```

## 🍁 Private Method 🌼

- to make method private double underscore in front of method name

```python
class Student:
    def __init__(self, name, degree):
        self.name = name                   
        self.__degree = degree
    
    def show(self):
        #accessing the private data member
        print("name", self.name, "degree", self.__degree)

    def __private_method(self):
        print("This is a private method")
```

```python
obj1 = Student("Alice", "Masters")
```

```python
obj1.private_method() # throw an error
```

```python
obj1.__private_method() # throw an error
```

- you can access private method by underscore class name and double underscore method name

```python
obj1._Student__private_method()
```

## ✨ Wrapper 🍁

- It is a way to provide an option to see the private method 

```python
class Student:
    def __init__(self, name, degree):
        self.name = name                   
        self.__degree = degree
    
    def show(self):
        #accessing the private data member
        print("name", self.name, "degree", self.__degree)

    def __private_method(self):
        print("This is a private method")

    def access__private_method(self):            # add a method to access private method
        self.__private_method()
```

```python
obj1 = Student("Alice", "master")
obj1.access__private_method()
```

## 🌼 Protected Variable 🍁  

- Within the class and its sub class, protected members can be access, (_)
- In industries you don't usually use protected variable  
- You are able to access a variable inside a class but if you are inherting something in case of inheriting base class the in those cases private variables will not be accessable so, protected variable comes into picture
- In this infornt of variable name give _ (single underscore) to make it Protected and also while accessing

```python
class College:
    def __init__(self):
        self._college_name = "DataAnalyst"
class Student(College):
    def __init__(self, name):
        self.name = name
        College.__init__(self)  #accessing variable of base class - Calssname__init__(self)

    def show(self):
        print("name", self.name, self._college_name) # directly call the base class variable using single underscore
```

```python
stud = Student("Alice")
```

```python
stud.name
```

```python
stud.show()
```

```python
coll = College()
coll._college_name   # accessing the protected variable
```

```python
# Another way of accessing protected variable

class College:
    def __init__(self):
        self._college_name = "DataAnalyst"
class Student(College):
    def __init__(self, name):  
        self.name = name
        super().__init__()       # super means main class College - include super and remove self

    def show(self):
        print("name", self.name, self._college_name) # directly call the base class variable using single underscore

```

```python
stud = Student("Alice")
```

```python
stud.show()
```
