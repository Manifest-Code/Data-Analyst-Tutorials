## 🍁 Property Decorators ✨ 

- Property decorators - you are allowed to use class method as an attribute

```python
class Student:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price
```

```python
#here name and price are private variables which you will not be able to acces from outside the class
stud = Student("Alice", 3000)
```

```python
stud.name #error as name and price private variables
```

```python
#still you can access private variable if you know structure of the class
stud._Student__name #using class name and exact variable name
```

```python
#another way to expose private variables using property decorators

class Student:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    @property
    def access_price(self):
        return self.__price
```

```python
stud = Student("Alice", 1000)
```

```python
stud.access_price
```

```python
# you want to modify the price - like set the price everytime new value is added

class Student:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    @property
    def access_price(self):
        return self.__price

    @access_price.setter    # once you call this method and set the price it will make self price as this new fixed price and then when you call access price method it will show set price value
    def price_set(self, new_price):
        self.__price = new_price
```

```python
stud = Student("Alice",1000)
```

```python
stud.access_price
```

```python
stud.price_set = 1500
```

```python
stud.access_price
```

```python
#delete a private variable

class Student:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    @property
    def access_price(self):
        return self.__price

    @access_price.setter    # once you call this method and set the price it will make self price as this new fixed price and then when you call access price method it will show set price value
    def set_price(self, new_price):
        self.__price = new_price

    @access_price.deleter
    def access_price(self):
        del self.__price
```

```python
stud = Student("Alice", 1000)
```

```python
stud.access_price
```

```python
stud.set_price = 1500
```

```python
stud.access_price
```

```python
del stud.access_price
```

```python
stud.access_price
```

```python
class Circle:
    def __init__(self, radius):
        self.__radius = radius
    @property
    def access_radius(self):
        return self.__radius
    @access_radius.setter
    def access_radius(self, value):
        self.__radius = value
    def area(self):
        return 3.14 * self.__radius*self.__radius
```

```python
rad = Circle(5)
```

```python
rad.access_radius
```

```python
rad.access_radius = 4
```

```python
rad.access_radius
```

```python
rad.area()
```
