# Single Inheritance - When a derived class has only one parent class  

![image.png](single_inheritance.png)

````python
class Father:
    def father_property(self):
        print("This is the father property")
````

````python
class Son(Father):
    def job(self):
        print("Son has this property")
````

````python
# child can have his property and also his father property
child_obj = Son() #object/instancce of Son class
````

````python
child_obj.job()
````

````python
child_obj.father_property()
````

````python
#A father will have only his property
father_obj = Father()
father_obj.father_property()
````

````python
father_obj = Father()
father_obj.job()         # parent can't access child property
````

## 🌼🪄 Method Overriding - redefining same method of parent class in child child/ derived class  

- if you create an object and call the same method it will print whatever is in the child class.  

- method overrding child class is very power  

- I happnes when there are only two classes one parent and one child (i.e. Single inheritance)

````python
# Method Overriding - redefining same method of parent class in child child/ derived class
class Fruit:
    def fruit_info(self):
        print("Inside parent class")
class Apple(Fruit):
    def fruit_info(self):                 # same method
        print("Inside the child class")
    def apple_info(self):
        print("Inside the child class")
````

````python
obj = Apple()
obj.fruit_info()    # if you create an object and call the same method it will print whatever is in the child class
````

## 🍁 Multi - Level Inheritance 🍀

![image.png](multi_level_inheritance.png)

````python
class Grandfather:
    def property_of_grandfather(self):
        print("I am your grand father, This is my property")
class Father(Grandfather):
    def property_of_father(self):
        print("I am your father, This is my property")
class Son(Father):
    def property_of_son(self):
        print("I am the king, I am having all the property")
````

````python
son = Son()       # son can access all the property of father and grandfather
son.property_of_father()
````

````python
son.property_of_grandfather()
````

````python
son.property_of_son()
````

````python
dad = Father()                 # father can access only his and his father (i.e grandfather's property) but can't access child's property
dad.property_of_father()
````

````python
dad.property_of_grandfather()
````

````python
dad.property_of_son()   # father can't access son proprty
````

````python
gf = Grandfather()             # Grand father can access only his propety and can't access his son or grandson's property
gf.property_of_grandfather()
````

````python
gf.property_of_father()
````

## 💎 Diamond Problem: ✨ 

- It occurs when a class inherits from 2 or more classes, it will lead to ambiguty in which method to execute.  

- Method will execute from parent 1 or parent 2 ?  

- To remove diamond problem - python uses method resolution order(MRO) algorithm called c3 linearization  

- In this the parent class which is inherited first in child class, that method will be executed.

````python
class Parent1:
    def parent1_info(self):                # method is same in both parents
        print("Inside parent1 class")
class Parent2:
    def parent1_info(self):                # same method
        print("Inside parent2 class")
class ChildClass(Parent1, Parent2):
     def child_info(self):               
        print("Inside child class")
````

````python
c2 = ChildClass()
c2.parent1_info()  #the parent class which is inherited first in child class, that method will be executed. in this case parent 1
````

````python
# Another example of diamond problem
class A:
    def method(self):                # all methods are same in parent class
        print("Method of class A")
class B(A):
    def method(self):                
        print("Method of class B")
class C(A):
     def method(self):                
        print("Method of class c")
class D(C, B):
    pass
````

````python
d = D()
d.method()    # it will print method of class c cause it is first inherited 
````

## 🪶 Abstraction 🎯  

- The concept of hiding complex implementation.  
- Hidding unneeded details and exposing the required only methods.
- Abstract class can not be instantiated (can not make object).
- abstract class should always be subclassed.
- abstract class may contain methods that are just declared and not implemented in abstract class itself, subclass are responsible for implimenting these methods.  
- If in inherited class a method is not present from abstract class it will not throw error.

## abstraction can be achived using abstrct class - import abc

````python
import abc # collection of multiple functions in a module
````

````python
#use case
class DataAnalyst:

    @abc.abstractmethod          # decoritum @abc, this methods have been declered but not implemented just passing
    def student_details(self):
        pass

    def student_assignment(self):
        pass

    def student_marks(self):
        pass
````

````python
class Python(DataAnalyst): # class inherited from DataAnalyst class
    def student_details(self):
        return "Data Science Course details"
    def student_marks(self):
        return "give ds student marks"
````

````python
class Oops(DataAnalyst):
    def student_assignment(self):
        return "This will give oops Assignment"
    def student_marks(self):
        return "oops student marks"
````

````python
pyth = Python()
````

````python
pyth.student_assignment()   # it will not throw any error even though method is not define
````

````python
pyth.student_details()
````

````python
p = Oops()
p.student_assignment()
````

````python
p.student_marks()
````

````python
# Use case
class Shape:

    @abc.abstractmethod
    def Calculate_area(self):
        pass

class Rectangle(Shape):
    def Calculate_area(self):
        return "area of rectable is length * breadth"
    
class Circle(Shape):
    def Calculate_area(self):
        return "area of circle is pi r**2"
````

````python
rect = Rectangle()
rect.Calculate_area()
````

````python
c = Circle()
c.Calculate_area()
````
