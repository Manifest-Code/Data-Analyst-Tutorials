# Single Inheritance - When a derived class has only one parent class  

![image.png](single_inheritance.png)

````python
class Father:
    def father_property(self):
        print("This is the father property")
````

````python
class Son(Father):
    def job(self):
        print("This is the father property")
````

````python
child_obj = Son() #object/instancce of Son class
````
