````python
# extract odd and even nos from a list:

class Listops:
    l = [1, 2, 3, 4, 5]
    def extract_even(self, l):
        l1 = []
        for i in l:
            if i % 2 == 0:
                l1.append(i)
        return l1
    def extract_odd(self, l):
        l1 = []
        for i in l:
            if i % 2 != 0:
                l1.append(i)
        return l1
````

````python
# now first extract the list from class
oops1 = Listops()
oops1.l
````

````python
oops1.extract_even(oops1.l)
````

````python
oops1.extract_odd(oops1.l)
````

````python
# changing the list
oops1.extract_odd([8, 11, 14, 13])
````

````python
# if you make any object the attribute l is fixed
# inspite of passing a different list above you are still getting the fixed l(1, 2, 3, 4, 5)
oops1.l
````

````python
# be careful as init has been used only pass self in the method and not the variable again

class Listops:
    def __init__(self, l):    
        self.l = l            
    def extract_even(self):   # be careful as init has been used only pass self in the method and not the variable again
        l = self.l            # add l = self.l
        l1 = []
        for i in l:
            if i % 2 == 0:
                l1.append(i)
        return l1
    def extract_odd(self):    # be careful as init has been used only pass self in the method and not the variable again
        l = self.l             # add l = self.l
        l1 = []
        for i in l:
            if i % 2 != 0:
                l1.append(i)
        return l1
````

````python
oops1 = Listops([1, 5, 11, 14])
oops1.extract_even()
````

````python
oops1 = Listops([1, 5, 11, 14])
oops1.extract_odd()
````

````python
class Listops:
    def __init__(self, l):    
        self.l2 = l             # you can also write any variable name like self.l2
    def extract_even(self):   
        l = self.l2            
        l1 = []
        for i in l:
            if i % 2 == 0:
                l1.append(i)
        return l1
    def extract_odd(self):    
        l = self.l2             
        l1 = []
        for i in l:
            if i % 2 != 0:
                l1.append(i)
        return l1
````

````python
oops1 = Listops([1, 3, 9, 11, 12])
oops1.extract_even()
````

````python
# Multiple attributes
#in a library a book is asociated with author, 
class Book:
    def __init__(self, name, author, title):
        self.name_of_books = name
        self.book_author = author
        self.title_name = title

    def extract_details_name_title(self):
        print(self.name_of_books, self.title_name)

    
    def extract_details_name_author(self):
        print(self.name_of_books, self.book_author)

````

````python
student1 = Book("ML", "Murphy", "Linear Regressions")
````

````python
student1.extract_details_name_author()
````

````python
student1.extract_details_name_title()
````

````python
student2 = Book("stats", "joss", "descriptive stats")
student2.extract_details_name_title()
````
