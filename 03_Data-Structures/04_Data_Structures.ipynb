 ##🌷 What is a Data Structure?

Data Structure is a way of organizing, storing, and managing data so that it can be accessed and modified efficiently.

🍁 Classification of Data Structures:  
   💎 Primitive – int, float, char, boolean    
   💎 Non-Primitive – array, list, stack, queue, tree, graph    
   💎 Linear – array, list, stack, queue    
   💎 Non-Linear – tree, graph

## Deep copy and shollow copy

## Shollow copy >> value will change with change in other list  
ex-

````python
grocery_list = ["apple", "banana", 1, 2.2, "kiwi", "mango"]
grocery_list
````

````python
a = ["A", "B"]
a = grocery_list # grocery list points towards the same memory location as a and vice versa
a
````

````python
grocery_list   # grocery list is same as earlier
````

````python
a[0] = "watermelon" 
a
````

````python
grocery_list # in grocery list at index 0 value changes as in a
````

## Deep Copy >> Will not change the value in another list as it will create another list in memory block

````python
grocery_list = ["apple", "banana", 1, 2.2, "kiwi", "mango"]
````

````python
b = grocery_list.copy() # in deep copy a will point to different memory location than grocery_list
````

````python
b
````

````python
b[0] = "grapes"
b
````

````python
grocery_list  # in deep copy grocery_list remains unchanged
````

## 📙 3. Tuple

## Definition:
 An ordered, immutable collection of elements.

## Key Characteristics:  
  💎 Cannot be modified  
  💎 Faster than list  
  💎 Allows duplicates  
  💎 Heterogeneous data 

  ex- aadhar card number, atm card number, position co-ordinate in maps

````python
t = (1, 2, 3)
t
````

````python
t[0]
````

````python
t[0] = 10  # This will raise an error because tuples are immutable
````

````python
# We can put tuples inside a tuple as well
box_1 = ("chocolate1", "chocolate2")
box_2 = ("chocolate3", "chocolate4")
choc_bag = (box_1, box_2)
choc_bag
````

````python
for box in choc_bag:
    for chocolate in box:
        print(chocolate)
````

## 📗 4. Set

## Definition:
 An unordered collection of unique elements.

## Key Characteristics:  
  💎 No duplicates  
  💎 Unordered  
  💎 Fast membership testing

Ex- where unique data is required , grocery list

````python
s = {1, 2, 3}
s
````

````python
type(s)
````

````python
# Sets do not allow duplicate values
s={"red", "black", "blue", "red", "green", "black"}
s
````

````python
# set have unordered elements, heterogeneous elements and no duplicate elements
set={"a", 1, 2.5, "b", 1, "a"}
set
````

````python
set[0]  # no point of indexing as sets are unordered and unique
````

````python
# you can not call set elements by index but can append it using add or remove elements

set.add("apple")
set
````

## 📕 5. Dictionary

## Definition:
 A collection of key–value pairs.

## Key Characteristics:  
  💎 Keys must be unique  
  💎 Fast lookup  
  💎 Unordered (insertion-ordered in Python 3.7+)  
  ex- contacts - (dad:number , mom:number)  , Restaurant dish:price

````python
d = {"name": "DataAnalyst", "age": 24}
d["name"]
````

````python
d = {"name": "DataAnalyst", "age": 24, "age": 30}
d["age"]  # will return 30 as keys are unique and last value will be assigned
````

````python
d.values()
````

  💎 Used in recursion, undo operations  

  ex-

````python
stack = []
stack.append(10)
stack.pop()
````

## 📗 7. Queue

## Definition:  
  A linear data structure that follows FIFO (First In, First Out).

## Key Characteristics:  
  💎 Operations: enqueue, dequeue  
  💎 Used in scheduling  

  ex-

````python
from collections import deque
q = deque()
q.append(10)
q.popleft()
````

## Key Characteristics:

  💎 No random access

  10 → 20 → 30 → NULL
