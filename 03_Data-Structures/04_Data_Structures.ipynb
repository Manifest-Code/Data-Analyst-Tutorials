## Queue - first in first out

![queue image.png](<attachment:queue image.png>)

````python
from collections import deque
# collection is a library and deque is a module
````

````python
checkout = deque()
checkout.append("Customer 1")
checkout.append("Customer 2")
checkout.append("Customer 3")
````

````python
checkout
````

````python
while checkout:
    customer = checkout.popleft()
    print("Serving", customer)
````

````python
from queue import Queue
print_queue = Queue()
print_queue.put("Document 1")
print_queue.put("Document 2")
print_queue.put("Document 3")

while not print_queue.empty():
    print_job = print_queue.get()
    print("Printing", print_job)
````

## Set Operations  

 💎 Union: Combines elements from two sets excluding duplicates  
 💎 Intersection: Only common elements from the set  
 💎 Difference: Returns the element that is present in 1st set and not in 2nd  
 💎 Symmetric difference: returns elements that are present in either of sets
    but not in both 

````python
s1 = {"hiking", "reading", "coding"}
s2 = {"coding", "traveling", "Photography"}
````

````python
s1
````

````python
s2
````

````python
# union >> combine all the elements
s1 | s2
````

````python
# Intersection >> common elements
s1 & s2
````

````python
# difference >> elements in s1 but not in s2
s1 - s2
````

````python
# symmetric difference >> elements in s1 and s2 but not in both
s1 ^ s2
````
