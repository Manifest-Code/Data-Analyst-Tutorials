- Positive Test Case → When everything is correct and the endpoint should behave as expected.
- Negative Test Case → When something is wrong or unexpected, and we check that the app handles it properly (e.g., wrong URL).

## Common HTTP Status Codes and Their Meanings  

```
   Status Code                Name                       Meaning
   200 OK                     Success                    The request was successful, and the server responded with the requested data.
   201 Created                Resource Created           A new resource was successfully created (e.g., after POST).
   400 Bad Request            Client Error               The server cannot process the request due to client error  
                                                         (e.g., malformed JSON, missing data). 
   401 Unauthorized           Authentication Required    The client must log in or provide a valid token/API key.
   403 Forbidden              Access Denied              The client is authenticated but not allowed to access this resource.
   404 Not Found              Resource Not Found         The requested URL/resource doesn’t exist on the server.
   405 Method Not Allowed     Wrong HTTP Method          The route exists, but doesn't support the HTTP method used 
                                                          (e.g., using POST on a GET-only route).
   422 Unprocessable Entity   Invalid Input Format       The server understands the request, but the input data is invalid or incorrect.
   500 Internal Server Error  Server Crashed             Something went wrong on the server while processing the request.
   503Service Unavailable     Server Overloaded          Server is down or busy; try again later

- create a app4.py file in that made changes to existing code 
- install a pytest package
- create a test_app3.py
- check testing the hello function
- write first positive test case for "/hello"  route test case in test_app4.py file
- then in bash terminal type pytest
- it will show 1 passed in 1._ _ s , collected item 1
- and also commented the failure test case uncomment it and if you run it also in bash terminal with typing pytest now it will run both passed and failure test cases

- Then wrote func test_predict_route_success()
- And tested with pytest
