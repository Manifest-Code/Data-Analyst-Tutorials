````python
## wrote the user input i.e name - {{name}} - to take user input in html have to write in {{}} (jinja2 template)
# create not a single text line but multiple line text

## import fask
from flask import Flask, render_template

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/', methods=["GET"])           
def home_page(): 
    return "Hellow World!"

## trigger the flask app
if __name__ == "__main__":
    app.run(debug = True, use_reloader = False)   
````

````python

## import fask
from flask import Flask, render_template

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/', methods = ["GET"])           
def form(): 
    return render_template("form.html")

@app.route('/form', methods = ["Post"])        #if you write get method here your information i.e name,password will be shown in url which is not secure thats why we use post method
def welcome():
    return "We have received your information"

## trigger the flask app
if __name__ == "__main__":
    app.run(debug = True, use_reloader = False)   
````
