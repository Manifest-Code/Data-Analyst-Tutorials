Create a index.html file. And consider html file as application 1 and this code as application 2. So we need to connect these two applications we can do that with the help of APIs.

```python
# create not a single text line but multiple line text

## import fask
from flask import Flask, render_template

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/')           
def home_page():  
    return render_template("index.html")

## trigger the flask app
if __name__ == "__main__":
    app.run(debug = True, use_reloader = False)   
```

```python
## wrote the user input i.e name - {{name}} - to take user input in html have to write in {{}} (jinja2 template)
# create not a single text line but multiple line text

## import fask
from flask import Flask, render_template

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/')           
def home_page(): 
    user_name = "Alice" 
    return render_template("index.html", name=user_name)

## trigger the flask app
if __name__ == "__main__":
    app.run(debug = True, use_reloader = False)   
```
