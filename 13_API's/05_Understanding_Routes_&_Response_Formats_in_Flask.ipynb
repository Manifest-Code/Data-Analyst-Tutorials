```python
# create base url '/' , '/about' , '/data'

## import fask
from flask import Flask

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/')           # expose flask app to the homepage i.e. route using decorator @app
def home():  
    return "Welcome to my Website"

@app.route('/about')
def about():
    data = "Data Analyst Course Lecture, Real World Examples of APIs"
    return data

@app.route('/data')
def data():
    user_Data = {"name": "Alice",
                 "Age": 22,
                 }
    return user_Data

## trigger the flask app
if __name__ == "__main__":
    app.run()  
```

```python
# create not a single text line but multiple line text

## import fask
from flask import Flask

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/')           # expose flask app to the homepage i.e. route using decorator @app
def home():  
    return "Welcome to my Website"

@app.route('/about')
def about():              # br next line  
    data = f"""<h1>About Us</h1><br>
              <h2>Data Analyst Lecture is about Real World Examples of APIs</h2>
              <h3>Thank you for wisiting our Website</h3>""" 
    return data

@app.route('/data')
def data():
    user_Data = {"name": "Alice",
                 "Age": 22,
                 }
    return user_Data

## trigger the flask app
if __name__ == "__main__":
    app.run(debug = True, use_reloader = False)   # debug = True will automatically run the code in terminal whenever there is changes 
                              # use_reloader=False is critical in notebooks to prevent SystemExit
```

```python
# Use Jsonify file (json file) - it is referred for data 

## import fask
from flask import Flask, jsonify

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/')           # expose flask app to the homepage i.e. route using decorator @app
def home():  
    return "Welcome to my Website"

@app.route('/about')
def about():
    data = "Data Analyst Course Lecture, Real World Examples of APIs"
    return data

@app.route('/data')
def data():
    user_Data = {"name": "Alice",
                 "Age": 22,
                 }
    return jsonify(user_Data)

## trigger the flask app
if __name__ == "__main__":
    app.run()  
```
