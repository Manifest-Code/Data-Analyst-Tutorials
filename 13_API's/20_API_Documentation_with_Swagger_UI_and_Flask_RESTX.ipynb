1. What is Swagger / OpenAPI?
Swagger(now known as OpenAPI) is a tool used to document APIs in a standard, readable format—and even lets you try the API from your browser.
Why Do We Need It?
When you build APIs (like /predict or /upload), others —like frontend developers, testers, or clients —need to know:
 - What endpoints exist?
 - What data to send?
 - What response to expect?  

Swagger creates an interactive UIthat answers all of this automatically.  

Example:  
You build an API like:  
@app.route('/predict', methods=['POST'])  

Without Swagger, users have to ask you:  
“What fields do I send? In what format? What will I get back?”  

With Swagger, users visit:  
http://localhost:5000/docs  
→ See your API endpoints, input format, descriptions  
→ Try sending requests live in the browser!

2. What is Flask-RESTX and Why It’s Useful?  

Flask-RESTX:  
- A Python extension for Flask
- Adds Swagger UI automatically
- Lets you define:
  - Endpoints
  - Input/output models
  - Descriptions
  - Validation

## Why Use Flask-RESTX?  
```
Without Flask-RESTX                  With Flask-RESTX
You write all docs by hand           Docs are generated for you
No input validation                  Validates requests automatically
Harder for others to use             Easy-to-use Swagger UI at /docs

3. How Swagger Helps Frontend Devs, Testers, and Clients  

Frontend Developers:  
 - Know exactly what data to send
 - Can test API before backend is "finished"
 - No guessing field names or formats  

QA/Testers:  
 - See what response to expect
 - Validate API behavior quickly  

Clients/Managers:  
 - Understand the service without reading code
 - Can try the API demo in a web browser

```
