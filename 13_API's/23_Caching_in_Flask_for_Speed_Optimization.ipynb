- It is the method were temporary memory is stored in temporary memory loaction.
- It is a layer between backend and frontend.
- when there are multiple users making a request it goes to backend and it has to process request and give same response again and again
- so to reduce a load on bakend server the cache layer is created in which you can temporary store a repetative respone 
- so when user makes request it create response and sends to user.
- if theres anything additional then only it sends request to backend

## Flask-Caching Library (Beginner-friendly)
- Install: pip install Flask-Caching
- Setup in app.py:
- from flask_caching import Cache
- app = Flask(__name__)
- cache = Cache(app, config={'CACHE_TYPE': 'SimpleCache'})

- Use in Blueprint route:
- @predict_bp.route('/predict', methods=['POST'])
- @cache.cached(timeout=30, query_string=True)
def predict(): …  

Now, for 30 seconds, the same input won’t hit the model again!

2. Avoid Reloading Models for Every Request  

```
Bad:  

with open('model.pkl') as f: 
   model = pickle.load(f) # every time!

Better: at top of the code write the model path to execute code faster
model = load_model_once() # load once on app start

3. Use Gzip Compression (Optional)  
when theres a large data - Compresses large JSON responses to improve speed.

4. Log Exceptions and Errors  
Use try-except blocks in routes and log errors for debugging.

5. Use a Production Server (e.g., Gunicorn)  
Flask's built-in server is for development only.  
Use Gunicorn or uWSGI + Nginx for real deployment.
