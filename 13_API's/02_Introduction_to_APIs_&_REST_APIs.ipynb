## Web API Concept

1. Client-Server Architecture: 
   - In a Web API, the client (usually a web application or mobile app) sends a request to the server, which processes the request and sends back a response.  
   - The Web API acts as the intermediary that handles this communication. 
 
 
2. End-to-End Interaction :  
 
   - API Call (Startpoint): The client sends a request to the server, specifying an operation (e.g., retrieving data)   
   - Endpoint: The specific URL where the API call is directed (Endpoint url consists of response given by server to client web)   
   - Response: The server processes the request and sends back the requested data or an error message.

## API Protocols/ Architechtal/ Methods: (SOAP, REST, GraphQL)  

1. SOAP (Simple Object Access Protocol):  
   - Definition: A protocol for exchanging structured information in web services using XML (XML/Html - Extensible Markup Langauges)
   - Characteristics: It is highly standardized, supports only XML format, and is often used in enterprise applications for secure, transaction-oriented operations 
   - Usage: Common in payment gateways, CRM solutions, and telecommunication services. 

2. REST (Representational State Transfer):  
   - Definition: An architectural style that uses standard HTTP methods (HTTP - Hyper text transfer protocol) (GET, POST, PUT, DELETE) for web services 
   - Characteristics: REST is lightweight, supports multiple formats (JSON, XML, etc.), and is widely adopted for public APIs. It treats data as resources, each identified by a unique URL 
   - Usage: REST is popular for building scalable web services, such as those provided by social media platforms, travel companies, and online marketplaces. 

3. GraphQL :  
   - Definition: A query language for APIs that allows clients to specify exactly what data they need 
   - Characteristics: Unlike REST, which returns a fixed structure, GraphQL enables clients to request precise data, making it efficient for applications with complex data requirements 
   - Usage: Ideal for modern web and mobile apps where data needs to be fetched from multiple sources in a single request.  
   - In this only one API is used to do multiple tasks/Database unlike in REST API for each task/Database there is different APIs

REST API (Representational State Transfer): 
 
  - REST (Representational State Transfer) is an architectural style that defines a set of constraints to be used for creating web services.

  - Web services that adhere to the REST architecture are called RESTful services or REST APIs. 

  - These APIs allow different software applications to communicate over the internet, often using standard HTTP methods. 


## Principles of REST API  

REST APIs are based on following key principles that guide their design and implementation :
 
**1. Client-Server Architecture:** The client and server have distinct responsibilities. The client handles the user interface and user-related tasks, while the server handles the data storage and processing   

**2. Statelessness:** Each request from a client to the server must contain all the necessary information for the server to fulfill that request. The server does not store any client context between requests, which enhances scalability and simplifies server design. 

        Stateful and Stateless analogy examples :  

        🔁 Stateful API — "The tea shop that remembers you"  
        You walk in.  
        The shopkeeper remembers you from last time: your name, your favorite tea, your sugar preference.  
        You just say, "The usual, please!" and they know what to do.  
        ✅ Key point: The shop remembers your past visits (your state).  

        - Also Like you login to website and close it and again login it doesn't ask you the password again. website opens directly.

        🔁 Stateless API — "The tea shop that forgets you every time"  
        You walk in.  
        The shopkeeper has no idea who you are.   
        Every time, you have to say: "Hi, I'm Alex. I want ginger tea with 1 spoon of sugar."  
        You have to give full details every time, because they don’t remember anything.  
        ✅ Key point: The shop doesn't keep track of your past (no state).  

        - when you login to website and close it and again login it ask you for the password again. website doesn't opens directly.

**3. Layered System:** The architecture allows for layers between the client and the server, such as caching and load balancing  

🧱 Layered System Analogy: Ordering Pizza 🍕
 
You want to order a pizza. Here's how the layered system looks:
 
You (Client) place an order through an app.
Delivery Call Center (API Layer) receives and routes your order.
Pizza Kitchen (Backend Server) prepares the pizza.
Ingredients Store (Database Layer) gives the dough, cheese, etc.
Delivery Guy (Network Layer) brings it to your home.
 
🔄 You (the client) only interact with the call center (API). You don’t care how the kitchen works or where they get ingredients. You just want your pizza.
That’s the layered system.



Layered System in REST Terms : 
```
 Layer                                 Role
 Client                           The user or app making the request (browser, mobile app, etc.)   
 API Gateway / Proxy              Forwards request, applies security, caching, or load balancing
 Server                           Processes the request, applies logic (backend logic)
 Database / Storage               Stores data that the server accesses and returns to the client


 
**4. Cacheability:** Responses from the server can be cached to improve performance, reducing the need for repeated requests and improving responsiveness.  
- Cache is a temporary memeory.
-  example: Suppose many users at a time selecting Nike shoes on flikart instead of hitting the server evrytime to get the same response the request goes to cache wgich holds temporary memory which gives the response to user

**5. Uniform Interface :** The Uniform Interface means that all REST APIs should follow standard rules for how resources are accessed, named, and represented. This includes using consistent API naming conventions, standard HTTP methods (like GET, POST, PUT, DELETE), and returning proper, well-structured responses (typically in JSON or XML)


## HTTP Methods (GET, POST, PUT, DELETE, etc.)

REST APIs use standard HTTP methods to perform operations on resources. 
Each method corresponds to a specific action:  
 
**POST:** Creates a new resource   
**GET:** Retrieves a resource or collection of resources   
**PUT:** Updates an existing resource   
**DELETE:** Remove a resource
