```
Production Ready Flask App Project Structure:  
```
        --my_flask_app/     
             --app/    
                --__init__.py   - Initialize flask app and extensions  
                -- routes.py    - Define URL routes and views  
                -- models.py    - Define database models (tables)  
                -- forms.py     - Web forms using flask-WTF (optional)  
                -- static/      - Static files(css, JS, Images)  
                -- templates/   - HTML templetes(Jinja2)  
                   --base.HTML  - Base layout templete  
        --config.py             -configuration settings  
        --run.py                 - Entry point to rin the app  
        --requirement.txt        - List of python packages  
        --ReadME.md              - project documentation  

        It's good for larger apps, teamwork, or production environments.

## Four Steps TO Build any FLASK APP:  

- import fask  
- create flask instances  
- define function and route  
- trigger the flask app  

````python
## import fask
from flask import Flask

## create flask instances 
app = Flask(__name__)  

## define function and route
@app.route('/')           # expose flask app to the homepage i.e. route using decorator @app
def home():  
    return "Hello World!"

## trigger the flask app
if __name__ == "__main__":
    app.run()  


## http://127.0.0.1:5000/  -HTTP -protocal, 127.0.0.1 -Local IP address, 5000 -port  #Default port where code is saved
````
