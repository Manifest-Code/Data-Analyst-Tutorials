## What is a Blueprint?
A Blueprintin Flask is like a mini Flask app inside your main app.  

It allows you to:  
  - Group related routes (like /predict, /user)
  - Keep code organized and modular
  - Reuse routes across apps if needed

## Real-World Analogy  
  - Think of Blueprints like folders in your computer —each folder contains related files.
  - So instead of dumping everything into app.py, you create folders like:  

```
routes/   
  ___ predict.py 
  ___ upload.py    

Each file registers its own Blueprint.  


## Steps to Use Blueprints  

1. Create a routes/ folder and a predict.py file inside it:  
routes/predict.py  

from flask import Blueprint, request, jsonify
import pickle
import pandas as pd  

predict_bp = Blueprint('predict', __name__)  
@predict_bp.route('/predict', methods=['POST'])  
def predict(): …..  

2. Register the Blueprint in app.py:  
app.py  

```
from flask import Flask  
from routes.predict import predict_bp  

app = Flask(__name__)  
app.register_blueprint(predict_bp)     

if __name__ == '__main__':   
   app.run(debug=True)  

Now /predict will work just like before —but your code is clean and modular.
