## Generator function

## regular function >> takes a list and gives square of each of the list  

```python
def square_number(n):
    result = []
    for i in n:
        result.append(i ** 2)
    return result

square_number([1, 4, 6]) 
```

```python
# regular function calculates the square of the values and returns in one go 
# scince its returning in one go >> memory usage will be more and 
# it will take lot of execution time if there is lot of computation as all the 
#computation will be done first and then the calculation will be returned in one go

def square_number(n):
    result = []
    for i in range(n):
        result.append(i ** 2)
    return result

square_number(10) 
```

```python
def square_number(n):
    result = []
    for i in range(n):
        result.append(i ** 2)
    return result

square_number(100000000000000000)   # If number is huge it takes time to generate result in one go
```

## can we have something to generate result one by one instead of one go
## yield >> generate function uses return statement but a generatoe function use yield statement

```python
def square_number_generators(n):
   for i in range(n):
        yield i ** 2
```

```python
square_number_generators(10)
```

```python
gen = square_number_generators(10)
gen
```

```python
next(gen)
```

## fabinocii series - 0, 1 are base after thar it add previous 2 digits

`0     1      1     2     3     5     8`
  
     `0+1    1+1   2+1   3+2`

```python
def fib(n):
    a=0
    b=1
    for i in range(n):
        yield a
        a, b = b, a+b
    
```
