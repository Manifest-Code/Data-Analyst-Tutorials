## Lambda function is also called anonymous function
## syntax
    lambda arguments: expression
    # no need to return statement and def keyword


```python
def square(x):
    return x**2
square(2)
```

```python
sq_lambda = lambda x: x**2
```

```python
sq_lambda(2)
```

```python
add = lambda a, b: a+b
```

```python
add(2, 5)
```

```python
is_even = lambda x: x%2 == 0
```

```python
is_even(4)
```

```python
x = ["Python", "Data Analyst", "Alice"]
x
```

```python
sorted(x)  # by default sorted based on alphabetical order
```

```python
# sort the list based on length of element of list
le = lambda x: len(x)
le("Alice")
```

```python
sorted(x, key = lambda x: len(x))
```

```python
# fabonachi series using lambda

fib = lambda n: n if n <=1 else fib(n-1) + fib(n-2)
[fib(i) for i in range(10)]
```

```python
# factorial of 5 

def fact_num(n):
    if n==0:
        return 1
    else:
        return n*fact_num(n-1)
```

```python
fact_num(5)
```

```python
fact_num = lambda n: 1 if n==0 else n*fact_num(n-1)
fact_num(5)
```
