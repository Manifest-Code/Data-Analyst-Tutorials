## Lambda function is also called anonymous function
## syntax
    lambda arguments: expression
    # no need to return statement and def keyword


```python
def square(x):
    return x**2
square(2)
```

```python
sq_lambda = lambda x: x**2
```

```python
sq_lambda(2)
```

```python
add = lambda a, b: a+b
```

```python
add(2, 5)
```

```python
is_even = lambda x: x%2 == 0
```

```python
is_even(4)
```

```python
x = ["Python", "Data Analyst", "Alice"]
x
```

```python
sorted(x)  # by default sorted based on alphabetical order
```

```python
# sort the list based on length of element of list
le = lambda x: len(x)
le("Alice")
```

```python
sorted(x, key = lambda x: len(x))
```

```python
# fabonachi series using lambda

fib = lambda n: n if n <=1 else fib(n-1) + fib(n-2)
[fib(i) for i in range(10)]
```

```python
# factorial of 5 

def fact_num(n):
    if n==0:
        return 1
    else:
        return n*fact_num(n-1)
```

```python
fact_num(5)
```

```python
fact_num = lambda n: 1 if n==0 else n*fact_num(n-1)
fact_num(5)
```

# Reduce >> folding / reduction  
# syntax  

  from functools import reduce   
  reduce(func, iterable)

## Reduce will always take 2 argument function

```python
from functools import reduce
```

```python
l = [2, 1, 3, 4, 4, 5, 6]
reduce(lambda x, y: x+y, l)
```

```python
l = [2, 1, 3, 4, 4, 5, 6]
reduce(lambda x, y, z: x+y+z, l)
```

```python
l = [2, 1, 3, 4, 4, 5, 6]
reduce(lambda x, y, z: x+y+z, [])
```

```python
words = ["Data", "science", "cources"]
reduce(lambda x,y: x+ " " +y, words)
```

```python
numbers = [1, 2, 5, 100, 1000, 6, 10000]
reduce(lambda x,y: x if x>y else y, numbers)
```

```python
def factorial(n):
    return reduce(lambda x,y: x*y, range(1, n+1))
```

```python
factorial(5)
```

## Filter Function:  

 It is used to filter elements from an iterable based on some condition  

 Syntax  
    
     filtter(function, iterable)


```python
l
```

```python
filter(lambda x: x%2 ==0, l)
```

```python
list(filter(lambda x: x%2 ==0, l))
```

```python
l1 = [-2, -1, 3, 4, 4, 5, 6]
list(filter(lambda x: x<0, l1)) # filter negative numbers
```

```python
list(filter(lambda x: x>0, l1)) # filter positive numbers
```

```python
s = ["ALex", "Data Analyst", "Data Science"]
list(filter(lambda x: len(x) > 4, s))
```

```python
def check_even(num):
    if num%2 == 0:
        return True
    return False
```
