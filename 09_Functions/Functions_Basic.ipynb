```python
# write a function that takes list as input and return all numeric values

def only_numeric(list):
    n=[]
    for i in list:
        if type(i) == int or type(i) == float:
            n.append(i)
    return n
```

```python
only_numeric([1, 4, 6.5, "Analyst"])
```

```python
# now in input take nested list:

def only_numeric(a):
    n=[]
    for i in a:
        if type(i) == list:
            for j in i:
                if type(j) == int or type(j) == float:
                    n.append(j)
        else:
            if type(i) == int or type(i) == float:
                n.append(i)
    return n
```

```python
only_numeric([3, 5, 5.5, True, [2, 4.5, "Analyst"]])
```

```python
# write a function can take any no. of values and return the list

def return_list(*arg):
    l = []
    for i in arg:
        if type(i) == list:
            l.append(i)
    return l
```

## Arbitrary Keyword Arguments(**kwarg):  

It allows function to accept any number of keyword arguments, which is collected in to dictionary

```python
def user_info(**args):
    for key, value in args.items():
        print(f"{key}:{value}")
```

```python
user_info(name="Alice", Age="20", city="New York")
```

## keyword only argument - you can force certain argument to be passed only by keyword by placing an asterisk * in the function signature. Any parameter after the * cannot be passed positionally. 

```python
def secure_delete(filename, * , force=False):
    print(f"Deleting {filename}, (force : {force})")
```

```python
secure_delete("data.txt", force=True)
```
