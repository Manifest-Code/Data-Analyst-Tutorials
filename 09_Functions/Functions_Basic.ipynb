- In Python, a function is a reusable block of code that performs a specific task. You can think of it as a "mini-program" inside your  
  main program that you can call whenever you need it.  
  ex- 

Inbuild functions: print, len, type, input etc

## Why the functions?  

- Functions are reusable.
- It keeps your code organiged

```python
print("Welcome to the office Alice")
print("Welcome to the office of Jack")
print("Welcome to the office of Mai")
```

```python
def greetings():
    print("Welcome to the office")
```

```python
greetings()
```

```python
def greetings(name): # passing a variable inside a function
    # positional argument: 'name'
    print("Welcome to the office", name)
```

```python
greetings("Alice")
```

```python
greetings("Jack")
```

```python
greetings("Mai")
```

```python
greetings()
```

```python
def greetings(Alice):
    print("Welcome to the office", Alice)
```

```python
greetings("Jack")
```

```python
greetings("Mai")
```

```python
def square_no(a): # a is called an argument. you can place anything here
    return a**2
```

```python
square_no(5)
```

```python
square_no(100)
```

```python
def square_no(args):
    return args**2
    
square_no(10)
```

```python
def addition(a, b):
    return a + b
```

```python
addition(5, 10)
```

```python
addition(2, 3)
```

```python
addition("Data","Analyst")
```

```python
addition([1, 2, 3], [4, 5, 6])
```

```python
addition("Data","Analyst") # as a, b are positional arguments that means they have fixed positions
```

```python
addition("Analyst", "Data") # here Analyst will be treated as 'a' and Data as 'b' 
```

```python
# a and b are positional arguments >> it maintains the order

# you want to fix above 'AnalystData' to 'DataAnalyst'

addition(b="Analyst", a="Data")  # by using keyword arguments you can change the order
```
